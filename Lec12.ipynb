# Python Exception Handling — Absolute Beginner to Advanced


---

## 0. Exception Handling kya hoti hai?

Program execute hote time agar koi **unexpected/problematic situation** aa jaye, Python ek **exception** raise kar sakta hai.

Example:

```python
a = 10
b = 0

print(a / b)
```

Output:

```text
ZeroDivisionError: division by zero
```

Yahan program normally aage nahi chalega. Python ne `ZeroDivisionError` naam ki exception raise ki.

### Simple definition

**Exception = runtime par aayi aisi abnormal condition jise Python signal karta hai aur jise hum handle kar sakte hain.**

Exception handling ka purpose:

- program ko unnecessarily crash hone se bachana
- error ko gracefully handle karna
- user ko meaningful message dena
- cleanup karna
- debugging information preserve karna
- program ke normal flow ko controlled way mein continue karna

---

# 1. Error vs Exception

Ye distinction important hai.

## Error

General sense mein error ka matlab hai program mein kuch wrong hona.

Examples:

- Syntax Error
- Runtime Error
- Logical Error

### Syntax Error

Code Python grammar follow nahi karta.

```python
if True
    print("Hello")
```

Python code execute karne se pehle hi complain karega:

```text
SyntaxError
```

Isko normal `try/except` se handle nahi kiya ja sakta because code parse hi nahi hua.

---

## Logical Error

Program run hota hai, crash bhi nahi karta, lekin result wrong hota hai.

```python
marks = 80
total = 50

percentage = marks / total * 10
```

Program execute ho jayega, but formula galat hai.

Logical errors ko exception handling solve nahi karti.

---

## Exception

Program syntactically valid hai, lekin execution ke during problem aa gayi.

```python
number = int("hello")
```

Result:

```text
ValueError
```

---

# 2. Python Exception ka basic flow

Jab exception aati hai, Python roughly ye process follow karta hai:

```text
Code execute
    ↓
Problem occur
    ↓
Exception object create
    ↓
Exception raise
    ↓
Python matching handler search karta hai
    ↓
Matching except mil gaya?
    ├── YES → handler execute
    └── NO  → exception propagate hoti hai
                  ↓
              caller ko jaati hai
                  ↓
              finally / cleanup
                  ↓
              unhandled hui → program traceback ke saath terminate
```

# 3. Exception object kya hota hai?

Exception sirf ek text message nahi hoti.

Python mein exception ek **object** hoti hai.

Example:

```python
try:
    10 / 0
except ZeroDivisionError as e:
    print(type(e))
    print(e)
```

Possible output:

```text
<class 'ZeroDivisionError'>
division by zero
```

Yahan `e` exception object ka reference hai.

Is object mein useful information ho sakti hai:

- exception type
- message
- traceback information
- cause/context
- custom attributes (custom exceptions mein)

# 4. `try` block

`try` ke andar wo code likhte hain jahan exception **expected** ho sakti hai.

```python
try:
    number = int(input("Enter number: "))
    print(100 / number)
```

Lekin sirf `try` likhna enough nahi hai. Usually uske saath `except`, `finally`, etc. use hota hai.

# 5. `except` block

`except` exception ko catch/handle karta hai.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("0 se divide nahi kar sakte.")
```

Output:

```text
0 se divide nahi kar sakte.
```

Important:

> `except` mein exception type specify karna usually better practice hai.

Avoid:

```python
try:
    risky_operation()
except:
    pass
```

Better:

```python
try:
    risky_operation()
except ValueError:
    print("Invalid value.")
```

# 9. Exception hierarchy

Python exceptions ek hierarchy follow karti hain.

Simplified structure:

```text
BaseException
├── KeyboardInterrupt
├── SystemExit
└── Exception
    ├── ArithmeticError
    │   ├── ZeroDivisionError
    │   └── OverflowError
    │
    ├── LookupError
    │   ├── IndexError
    │   └── KeyError
    │
    ├── OSError
    │

    │
    ├── TypeError
    ├── ValueError
    ├── AttributeError
    ├── NameError
    └── ...
```

Most application-level exceptions `Exception` ke descendants hoti hain.

# 10. `Exception` ko catch karna


But broad catch ko carefully use karna chahiye.

# 11. `BaseException` kyun generally catch nahi karna chahiye?

`BaseException` hierarchy ke top par hai.

Iske important direct descendants mein:

- `Exception`
- `KeyboardInterrupt`
- `SystemExit`
- `GeneratorExit`

hote hain.

Agar:

```python
except BaseException:
    pass
```

likh diya, to `Ctrl+C` jaise interrupts bhi accidentally swallow ho sakte hain.

Generally application code mein:

```python
except Exception:
```

preferred hai.

# 13. `else` with exception handling

`try-except` ke saath `else` bhi use kar sakte hain.

```python
try:
    number = int(input("Enter number: "))
except ValueError:
    print("Invalid input.")
else:
    print("Successfully converted:", number)
```

### Golden rule

`else` **tab execute hota hai jab `try` successfully complete ho.**

Agar `try` mein exception aayi → `else` skip.

# 14. `finally`

`finally` block almost always cleanup ke liye use hota hai.

```python
try:
    print("Working...")
except Exception:
    print("Error")
finally:
    print("Cleanup")
```

`finally` ka important characteristic:

> Exception aaye ya na aaye, `finally` normally execute hota hai.

---

# 15. `try-except-else-finally`

Full structure:

```python
try:
    risky_operation()

except SomeException:
    handle_error()

else:
    operation_successful()

finally:
    cleanup()
```

### Execution cases

### Case 1 — No exception

```text
try
 ↓
else
 ↓
finally
```

### Case 2 — Exception handled

```text
try
 ↓
except
 ↓
finally
```

### Case 3 — Exception unhandled

```text
try
 ↓
finally
 ↓
exception propagates
```

# 21. Custom exception mein data store karna

```python
class InsufficientBalanceError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount

        super().__init__(
            f"Balance={balance}, requested={amount}"
        )
```

Use:

```python
raise InsufficientBalanceError(500, 1000)
```

Handler:

```python
except InsufficientBalanceError as e:
    print(e.balance)
    print(e.amount)
```

# 23. Exception matching aur order

Ye important hai.

Wrong:

```python
try:
    ...
except Exception:
    ...
except ValueError:
    ...
```

`ValueError` kabhi second handler tak nahi pahuchega because `Exception` already usko catch kar lega.

Correct:

```python
try:
    ...
except ValueError:
    ...
except Exception:
    ...
```

### Rule

**Specific exceptions → pehle**  
**General exceptions → baad mein**

# 24. Bare `except`

Possible:

```python
try:
    ...
except:
    ...
```

But generally avoid.

Reason:

- bugs hide ho sakte hain
- `KeyboardInterrupt` catch ho sakta hai
- debugging difficult ho sakti hai
- unexpected failures silently swallow ho sakti hain

Better:

```python
except Exception as e:
    ...
```

Aur best jab possible ho:

```python
except ValueError:
    ...
```

# 30. Exception propagation

Agar current function exception handle nahi karta, exception caller ko propagate hoti hai.

```python
def divide():
    return 10 / 0

def calculate():
    return divide()

try:
    calculate()
except ZeroDivisionError:
    print("Handled")
```

Flow:

```text
divide()
 ↓
exception
 ↓
calculate()
 ↓
caller
 ↓
except
```

# 31. Function ke andar handle karna vs bahar

### Inside

```python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None
```

### Outside

```python
def divide(a, b):
    return a / b

try:
    result = divide(10, 0)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

Kaunsa better hai?

Context par depend karta hai.

General principle:

> Exception ko us level par handle karo jahan tumhare paas actually meaningful recovery/action possible ho.

# 32. EAFP vs LBYL

Python mein famous philosophy:

## EAFP

**Easier to Ask Forgiveness than Permission**

Try karo, exception handle karo.

```python
try:
    value = data["name"]
except KeyError:
    value = "Unknown"
```

## LBYL

**Look Before You Leap**

Pehle check karo.

```python
if "name" in data:
    value = data["name"]
else:
    value = "Unknown"
```

Python often EAFP style prefer karta hai, but blindly nahi.

# 33. Exception chaining

Suppose low-level error hua:

```python
try:
    number = int("abc")
except ValueError as e:
    raise RuntimeError("Configuration invalid") from e
```

Yahan original `ValueError` ko new `RuntimeError` ka explicit cause banaya gaya.

Ye:

```python
raise ... from e
```

**explicit exception chaining** hai.

---

# 34. Why chaining useful hai?

Large application mein layers hoti hain:

```text
Database
   ↓
Repository
   ↓
Service
   ↓
API
```

Database-specific exception ko directly API user ko nahi dikhana chahte.

Instead:

```python
try:
    load_user()
except DatabaseError as e:
    raise UserRepositoryError("Unable to load user") from e
```

Isse abstraction maintain hoti hai aur original cause bhi preserve hota hai.

- propagate karna hai?

# 51. Retry pattern

Temporary failures ke liye retry useful ho sakta hai.

Basic idea:

```python
for attempt in range(3):
    try:
        result = risky_operation()
        break
    except TemporaryError:
        if attempt == 2:
            raise
```

Real systems mein:

- exponential backoff
- jitter
- maximum attempts
- idempotency

consider karna chahiye.

# 62. Exception handling with threads

Thread ke andar exception automatically main thread ke `try-except` mein nahi aa jaati.

Example concept:

```python
def worker():
    raise ValueError("Worker failed")
```

Agar thread ke andar handle nahi kiya gaya, behavior thread execution framework par depend karega.

`concurrent.futures` ka `Future` model exception ko caller tak re-raise karne ka convenient mechanism deta hai.

---

# 63. `Future.result()`

Conceptual:

```python
from concurrent.futures import ThreadPoolExecutor

with ThreadPoolExecutor() as executor:
    future = executor.submit(risky_operation)

    try:
        result = future.result()
    except Exception as e:
        print("Worker failed:", e)
```

`future.result()` worker ki exception ko caller context mein raise kar sakta hai.

```python
errors = [
```

# 65. `except*`

`ExceptionGroup` ke saath `except*` use hota hai.

```python
try:
    raise ExceptionGroup(
        "errors",
        [
            ValueError("bad value"),
            TypeError("bad type")
        ]
    )
except* ValueError as eg:
    print("Value errors:", eg)
except* TypeError as eg:
    print("Type errors:", eg)
```

Important:

`except` aur `except*` same concept nahi hain.

- `except` normal exception handling ke liye
- `except*` exception groups ke individual matching parts handle karne ke liye

# 70. Don't use exceptions for normal control flow unnecessarily

Bad:

```python
try:
    value = data["name"]
except KeyError:
    value = None
```

This may be fine in some Pythonic situations.

But if you simply need a default:

```python
value = data.get("name")
```

Often cleaner.

Exception handling should represent exceptional/alternative conditions, not every normal branch.

# 71. Don't catch what you cannot handle

Important principle:

```python
try:
    operation()
except Exception:
    pass
```

is usually bad.

Agar tum error recover nahi kar sakte, allow it to propagate or handle it at a higher meaningful layer.

Isse root cause preserve hota hai.

# 77. Advanced: `finally` and `return`

Ye tricky hai.

```python
def test():
    try:
        return 10
    finally:
        return 20
```

Result:

```text
20
```

Why?

`finally` ka `return` previous pending return ko override kar sakta hai.

**Best practice: `finally` mein return avoid karo.**

Similarly, `finally` mein exception ko accidentally suppress/replace karne wale control-flow constructs avoid karo.

# 78. Advanced: `finally` and exception

```python
def test():
    try:
        raise ValueError("Original")
    finally:
        print("Cleanup")
```

Output mein cleanup ke baad original exception propagate hoti hai.

But agar `finally` ke andar new exception raise kar di:

```python
def test():
    try:
        raise ValueError("Original")
    finally:
        raise RuntimeError("Cleanup failed")
```

Then original exception overshadow ho sakti hai.

Isliye `finally` mein risky operations carefully design karo.
