# Operators

```python
#Arithmetic Operators

print(5+6)   #Addition
print(5-6)   #Subtraction
print(5/2)   #Division
print(5//2)  #floor Division
print(5%2)   #Modulus operator (gives remainder)
print((5**2))#Power of operator(5^2)
```

```python
# Relational Operators 

print(5>2)

print(5<2)

print(5==2)

print(5>=5)

print(5<=5)

print(5!=2) 
```

```python
#Logical Operators 

print(1 and 0)

print( not 0)

print(1 or 0)
```

```python
#Bitwise Operators 
#Special operators jo binary values ke upar operate karte hai 

print(2&3) #Bitwise and (2:010 ,3:101,now both values are compaed means har lsb to msb ke beech and lgta hai jo final answer aata hai wohi print hojata hai )

print(2|3) #Bitwise or\

print(2^3) #Bitwise xor

print(~3) #Bitwise not
#HOw does it working now we know ki 8 bit ka storage hota hai to pehel 3 ka binary would be (0000 0011) then isko invert karenge (1111 1100) now sign bit is 1 yni negative number so  ab computer iska 2's complement nikalega so what we get is (0000  0011 +1)=4 and sign bit was 1 so output is -4

print(4>>3) #Bitwise right shift operator (4 ka binary likha and right side 2-2 se shift kiya )
#Working: 4 ko 2^2 se divide karo  that is 4//4 =1 
#also if 4>>3 then 4//(2^3)=4//8=0

print(4<<2) #BItwise left shift operator (4 ko 2^2 se  multiply karo that is 4 *4=16)
```

# Loops

```python
#where loops are used?
#In differennt softwares like in flipkart  when we search smartphones to kuch definite number of smartphones dikhte hai to similarities is like left main image then right main contents to yhi cheez baar baar repeat ho rhe hai h=therefore baar baar baane se acha  ek baar banao and baar baar db se fetch karke dikhado


#2 types:while,for

#Syntax
number=int(input("Enter your number "))
i=1
fac=1
while i<=number:
     fac*=i
     i+=1
print(fac)     
```

```python
#while loop with else

x=1
while x<3:
    print(x)
    x+=1
else:
    print("Limits crossed")    
```

```python
#Guessing games using loops 
import random as rnd
jackpot=rnd.randint(1,100)
count=0
user_guess=int(input("Enter your guess"))
while user_guess!=jackpot:

    if(user_guess<jackpot):
            print("guess higher")
            
    else:
          print("guess lower")
    user_guess=int(input("enter again"))
    count+=1     
else:
      print("Chalo akhir kaar guess kar hi liya,took you ",count," tries")
```

```python
#for loops 
for i in range(1,11):
    print(i,"\n")

#also you can also provide step size structure is somewhat like this:range(start,end,step_size)
for i in range(1,11,2):
    print(i)    
```
