# Iterators & Iterables
 
## Iteration Kya Hai?
 
Iteration ek general term hai — kisi bhi cheez ke har item ko ek-ek karke access karna aur possibly uske upar koi operation perform karna.
 
Simple words mein: loop chalate ho list ko access karne ke liye. 5 items hain to pehle pe gaye, access kiya, jo karna tha kiya, aage badh gaye — aisa hi har ek item ke saath hota hai.
 
```python
num = [1,2,3]
for i in num:
    print(i, end=" ")
# list ka har element access karwa diya through iteration
```
 
## Iterator Kya Hai?
 
Iterator ek **object** hota hai jo programmer ko allow karta hai kisi sequence of data ke upar go through karne mein, aur aisa karte hue woh entire data ko memory mein store nahi karta.
 
**Example:** Maan lo mujhe 1 se 1 lakh tak har number ko 2 se multiply karke dikhana hai — iske do tareeke honge:
 
```python
# Tareeka 1 — poori list bana lo
l = [x for x in range(1,100001)]
 
import sys
print(sys.getsizeof(l)/(1024**2))   # size MB mein — bytes se MB banane ke liye 1024² se divide
# Kuch 80 MB tak ka space le leta hai (tumhare system pe kam-zyada ho sakta hai)
 
# Tareeka 2 — iterator use karo
x = range(1,100001)
print(sys.getsizeof(x)/(1024**2))   # bahut hi kam space lega
```

## Iterable Kya Hai?
 
Iterable — aisa koi bhi object jiske upar loop chala sako.
 
Iterable, `iter()` method ko pass karne par ek **iterator generate karta hai**.
 
```python
l = [1,2,3]
# l ek iterable hai
 
type(iter(l))   # iterable ko iter() pass karne se ek iterator ban jaata hai
```
 
## Points To Remember
 
- **Every iterator is an iterable.**
- **But not all iterables are iterators.**
Bohot confusing lagta hai, pehle dusre point pe focus karte hain:
 
```python
l = [x for x in range(1,100001)]
```
Yeh ek **iterable** to hai (kyunki iske upar iteration kar sakte ho), par **iterator nahi hai** — kyunki yeh saare numbers ek saath memory mein load kar raha hai, ek-ek karke nahi (matlab ek ko load kiya, kaam kiya, hataya, phir dusra laaya — waisa nahi kar raha).
 
Point 1 kehta hai ki iterator ke upar loop chala sakte ho. Notice karo — `x = range(1,100001)` ek baar mein memory mein har number load nahi karta. Yeh sirf yaad rakhta hai ki starting 1 hai. Jab iske upar (jo ki iterator hai) loop chalta hai, to yeh starting value de deta hai, kaam hoke woh memory se hat jaata hai, phir jab loop wapas aata hai to yeh sochta hai "pehle 1 diya tha, ab iske aage ka dena padega" — to woh de deta hai. Isi wajah se yeh ek iterator hone ke bawjood khud pe iteration bhi karwa raha hai — matlab yeh ek iterable bhi hua.

## Kaise Pata Karein — Iterable Hai Ya Nahi?
 
Agar pata lagana hai ki koi cheez iterable hai ya nahi, to `dir()` mein us cheez ko daal do — agar uske andar `__iter__` method hai to woh iterable hai, warna nahi hai.
 
```python
l = [1,2,3]
dir(iter(l))
```
 
## Kaise Pata Karein — Iterator Hai Ya Nahi?
 
Yeh pata lagane ke liye bhi `dir()` ke andar daalo — agar `__iter__` aur `__next__` dono methods hain, to woh iterator hai, warna nahi.
 
```python
l = [1,2,3]   # l ek iterable hai, par kya yeh iterator hai?
 
dir(l)          # sirf __iter__ hai, __next__ nahi — matlab iterable hai, iterator nahi
dir(iter(l))    # iter(l) se l ka iterator banta hai — ismein __iter__ aur __next__ dono hain
```
 
## For Loop Internally Kaam Kaise Karta Hai?
 
Sabse pehle humare paas ek iterable hona chahiye. Phir Python loop encounter karne par uska **iterator fetch** karta hai, aur `next()` ko tab tak call karta hai jab tak data khatam nahi ho jaata.
 
```python
num = [1,2,3]
 
fetch = iter(num)   # iterator fetch kiya
next(fetch)          # 1
next(fetch)          # 2
next(fetch)          # 3
```

 
**Apna khud ka for loop bana ke dekhte hain:**
```python
def my_loop(iterable):
    iterator = iter(iterable)
    while True:
        try:
            print(next(iterator))
        except StopIteration:
            break
```
 
## Ek Confusing Concept
 
```python
num = [1,2,3]
iterator1 = iter(num)
print("Address of first iterator:", id(iterator1))
 
# iterator1 hamare iterable (num) ka iterator hai, par hum jaante hain
# ki iterator khud bhi ek iterable hota hai — to agar isi iterator pe
# iter() chala diya jaaye to kya hoga?
 
iterator2 = iter(iterator1)
print("Address of second iterator:", id(iterator2))
 
# iterator2 ban gaya, iske pass bhi __iter__ aur __next__ methods hain,
# aur iska address bhi iterator1 ke address ke bilkul same hai —
# matlab reality mein iterator1 aur iterator2 same object hain
```
Matlab: kisi iterator ke upar `iter()` lagane se koi **naya** object nahi banta — woh khud ka hi reference return kar deta hai (`__iter__` method usually `return self` karta hai).
 
## Apna Khud Ka `range()` Function Banana
 
Iterator protocol samajhne ke liye apni khud ki `range` jaisi class banate hain — iske liye do classes chahiye: ek **iterable class** (jo `__iter__` implement kare) aur ek **iterator class** (jo `__iter__` + `__next__` dono implement kare).
 
```python
class myownrange:
    def __init__(self, start, end):
        self.start = start
        self.end = end
 
    def __iter__(self):
        return iterator_of_my_range(self)
 
class iterator_of_my_range:
    def __init__(self, Iterable_object):
        self.iterator = Iterable_object
 
    def __iter__(self):
        return self
 
    def __next__(self):
        if self.iterator.start >= self.iterator.end:
            raise StopIteration
 
        current = self.iterator.start
        self.iterator.start += 1
        return current
 
for i in myownrange(1,11):
    print(i)
```

```mermaid
    C -->|"start >= end ho jaane par"| D["StopIteration raise<br/>loop khatam"]
```
 
## Iterators Kahan Use Hote Hain?
 
Jahan **stream of data** ho, wahan us poore data ko ek saath load kiye bina hi kaam ban jaata hai.
 
Example: Deep learning mein bahut saari images hoti hain, unhe arrays mein convert karke kuch operations karne hote hain. Ab agar images 20 GB ki hain to poora data ek saath memory mein kaise convert karoge? — Iterator ka use karke, ek-ek karke, bina poora data memory mein rakhe.
