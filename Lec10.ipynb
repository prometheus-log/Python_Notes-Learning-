# Python OOP: Abstraction (Beginner to Advanced)

**Abstraction** ka simple matlab hota hai: **"Hiding internal implementation details and showing only essential features to the user."** (Yaani, piche kaam kaise ho raha hai use chupana, aur bahar sirf wahi dikhana jo kaam ka hai).

### Real-World Example:
Jab aap **Car chalate ho**, toh aapko sirf Accelerator, Brake, aur Steering Wheel se matlab hota hai. Engine ke andar piston kaise upar-neeche ho raha hai ya fuel kaise pump ho raha hai, uski internal detail aapse chhupi (abstracted) hoti hai.

---

## 1. Encapsulation vs Abstraction: Dono me Kya Difference Hai?

Bohot se log in dono me confuse hote hain. Inhe simple language me samajhte hain:

| Feature | Abstraction (What it does) | Encapsulation (How it hides) |
| :--- | :--- | :--- |
| **Focus** | Yeh **"What"** par focus karta hai (Design level par ki object kya-kya karega). | Yeh **"How"** par focus karta hai (Implementation level par data ko kaise safe rakhna hai). |
| **Kaise hota hai?** | Yeh **Abstract Classes** aur **Methods** ke zariye hota hai. | Yeh **Private variables (`__`)** aur Getter/Setter ke zariye hota hai. |
| **Goal** | Iska goal complex system ko asaan banana (complexity kam karna) hai. | Iska goal data ko unauthorized access se protect karna (security) hai. |

## 2. Basic Concept: Abstract Base Classes (ABC) (Intermediate)

Python me default roop se pure abstract classes nahi hoti hain. Iske liye hume Python ke built-in module **`abc`** ka use karna padta hai.

### Key Rules of Abstraction:
1. Aap kisi bhi **Abstract Class ka object (instance) nahi bana sakte**.
2. Agar kisi class me ek bhi `@abstractmethod` hai, toh woh class Abstract Class ban jati hai.
3. Jo bhi Child class is Abstract class ko inherit karegi, use **saare abstract methods ko compulsory implement (define) karna padega**, nahi toh Python error de dega.

### Syntax Aur Basic Code:
```python
from abc import ABC, abstractmethod

# Yeh hamari Abstract Class hai (Blueprint)
class Vehicle(ABC):
    
    @abstractmethod
    def start_engine(self):
        pass  # Abstract methods me koi body nahi hoti, sirf 'pass' likhte hain

    @abstractmethod
    def stop_engine(self):
        pass

# Child Class 1
class Car(Vehicle):
    def start_engine(self):
        print("Car Engine Started: Vroom Vroom!")
        
    def stop_engine(self):
        print("Car Engine Stopped.")

# If you try to make an object of Vehicle:
# my_vehicle = Vehicle()  --> TypeError: Can't instantiate abstract class

my_car = Car()
my_car.start_engine()  # Output: Car Engine Started: Vroom Vroom!
```

## 3. Abstract Class with Normal Methods & Constructors (Advanced)

Ek abstract class me sirf abstract methods hi nahi hote. Usme **normal methods (concrete methods)** aur **`__init__` constructor** bhi ho sakte hain. 

```python
from abc import ABC, abstractmethod

class DatabaseConnector(ABC):
    def __init__(self, db_name):
        self.db_name = db_name  # Abstract class ka apna constructor

    # 1. Concrete Method (Sabhi childs ke liye common logic)
    def connection_log(self):
        print(f"Attempting to connect to database: {self.db_name}")

    # 2. Abstract Method (Har child apna custom logic likhega)
    @abstractmethod
    def connect(self):
        pass

class MySQLConnector(DatabaseConnector):
    def connect(self):
        self.connection_log() # Parent ka normal method use kiya
        print("MySQL Database connected successfully.")

# Usage
db = MySQLConnector("ECommerce_DB")
db.connect()
# Output:
# Attempting to connect to database: ECommerce_DB
# MySQL Database connected successfully.
```

## 4. Minute Technical Properties of Abstraction (Expert Details)

Aapko interview aur advanced levels par yeh choti-choti baten pata honi chahiye:

### A. Abstract Properties
Jaise aap `@abstractmethod` banate hain, waise hi aap properties ko bhi abstract bana sakte hain taaki child class majboori me un variables ke liye getters/setters likhe.
```python
from abc import ABC, abstractmethod

class Model(ABC):
    @property
    @abstractmethod
    def table_name(self):
        pass

class UserModel(Model):
    @property
    def table_name(self):
        return "users_table"
```

### B. Can Abstract Methods have code?
**Haan!** Yeh bohot kam log jaante hain. Python me abstract method ke andar bhi code likha ja sakta hai. Child class use `super().method_name()` se call karke extend kar sakti hai.
```python
from abc import ABC, abstractmethod

class BaseAPI(ABC):
    @abstractmethod
    def authenticate(self):
        print("Base Validation: Checking API tokens...") # Abstract method with body

class UserAPI(BaseAPI):
    def authenticate(self):
        super().authenticate() # Parent ka validation chalaya
        print("User Validation: Checking user permissions...")

u = UserAPI()
u.authenticate()
```

### C. Abstract Magic Methods
Aap python ke magic methods (jaise `__str__`, `__repr__`, `__len__`) ko bhi `@abstractmethod` bana sakte hain taaki child class structural guidelines ko break na kare.

## 5. Real-World Architectural Use Case

Project design karte waqt Abstraction ka asli use **API Integration** ya **Payment Gateways** me hota hai, jahan backend developers ek strict contract tayyar karte hain:

```python
from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    @abstractmethod
    def validate_user(self, user_id): pass

    @abstractmethod
    def debit_amount(self, amount): pass

    @abstractmethod
    def verify_status(self): pass

# Kal ko agar naya payment option (jaise Stripe) add karna ho, 
# toh use in teeno methods ko implement karna hi padega.
class Razorpay(PaymentGateway):
    def validate_user(self, user_id): print("Razorpay: User Validated")
    def debit_amount(self, amount): print(f"Razorpay: ₹{amount} Debited")
    def verify_status(self): print("Razorpay: Payment Success")
```
