## Part 1: Solving the Geometry Assignment (Image Problems)

Lekchar ki screen par diye gaye saare 2D Geometry ke scenarios ko handle karne ke liye hum do main classes banayenge: **`Point`** aur **`Line`**.

### Math Formulas Used:
1. **Distance between 2 points \((x_1, y_1)\) and \((x_2, y_2)\)**: \(\sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}\)
2. **Distance from Origin \((0, 0)\)**: \(\sqrt{x^2 + y^2}\)
3. **Line Equation**: \(Ax + By + C = 0\)
4. **Point lies on line**: Agar \(A(x_1) + B(y_1) + C == 0\) ho jaye.
5. **Distance between Point and Line**: \(\frac{\vert{}A(x_1) + B(y_1) + C\vert{}}{\sqrt{A^2 + B^2}}\)

### Implementation Code:

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # 1. User can view 2D coordinates
    def __str__(self):
        return f"Point({self.x}, {self.y})"

    # 2. Distance between 2 coordinates
    def distance_between(self, other_point):
        if isinstance(other_point, Point):
            return math.sqrt((other_point.x - self.x)**2 + (other_point.y - self.y)**2)
        return NotImplemented

    # 3. Distance of a coordinate from origin
    def distance_from_origin(self):
        return math.sqrt(self.x**2 + self.y**2)


class Line:
    # Line represented as Ax + By + C = 0
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def __str__(self):
        return f"{self.a}x + {self.b}y + {self.c} = 0"

    # 4. Check if a point lies on a given line
    def contains_point(self, point):
        # Ax + By + C should equal 0
        result = (self.a * point.x) + (self.b * point.y) + self.c
        return result == 0

    # 5. Distance between a given 2D point and a given line
    def distance_from_point(self, point):
        numerator = abs((self.a * point.x) + (self.b * point.y) + self.c)
        denominator = math.sqrt(self.a**2 + self.b**2)
        return numerator / denominator


# --- Execution and Testing ---
p1 = Point(0, 0)
p2 = Point(3, 4)
print(p2)                               # Output: Point(3, 4)
print("Distance:", p1.distance_between(p2))  # Output: 5.0
print("From Origin:", p2.distance_from_origin()) # Output: 5.0

# Creating a line: 3x - 4y + 0 = 0
line1 = Line(3, -4, 0)
p3 = Point(4, 3) # Is it on line? 3(4) - 4(3) = 12 - 12 = 0 (Yes)
print("Point lies on line?:", line1.contains_point(p3)) # Output: True

p4 = Point(2, 5)
print("Shortest distance to line:", line1.distance_from_point(p4))
```

## Part 2: Object Memory, Reference Variables & Mutability

Lekchar ke in important internal core concepts ko detail mein samajhte hain:

### 1. How Objects Access Attributes & Attribute Creation from Outside
Python mein jab aap `obj.name = "Rahul"` likhte ho, toh object apne internal dictionary (`__dict__`) mein check karta hai.
* **Superpower/Risk:** Python dynamic hai! Aap class ke bahar se bhi kisi bhi time naya variable kisi object mein chipka sakte ho (e.g., `p1.z = 10`), lekin yeh bad practice maani jaati hai kyunki baaki objects ke paas woh variable nahi hoga.

### 2. Reference Variables & Pass by Reference
* Jab aap `p1 = Point(3, 4)` karte ho, toh `p1` actual object nahi hai! Woh bas memory mein baney object ka **Address (Reference)** hold karta hai.
* Agar aap `p2 = p1` karoge, toh koi naya object nahi banega, dono variables ek hi memory location ko point karenge.
* **Pass by Reference:** Jab aap kisi object ko function ke andar pass karte ho, toh uska address pass hota hai. Function ke andar kiya gaya koi bhi change original object par asar dalega.

### 3. Mutability of Objects
Python ke custom user-defined objects **Mutable** hote hain. Iska matlab hai ki memory address badle bina aap unke andar ki values ko directly change kar sakte ho.

```python
# Pass by reference and Mutability Example
```

## Part 3: Encapsulation (Beginner to Advanced)

### What is Encapsulation?
Encapsulation ka matlab hai data (attributes) aur logic (methods) ko ek single unit/capsule ke andar wrap karna aur **Data Hiding** implement karna taaki bahar ka koi galat use na kar sake.

### Implementation Levels:
* **Public (`self.var`):** Kahin se bhi modify ho sakta hai.
* **Private (`self.__var`):** Sirf class ke andar ke methods access kar sakte hain. Python iska naam internally change kar deta hai (`_ClassName__var`), jise **Name Mangling** bolte hain.

### Advanced Getter and Setter Pattern:
Direct variables ko expose karne ke bajaye hum `@property` decorator use karte hain safe modification ke liye.

```python
class SecureAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # Private Attribute

    # Getter: Safely reading data
    @property
    def balance(self):
        return self.__balance

    # Setter: Validating before changing data
    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Error: Balance negative nahi ho sakta!")

# Usage
acc = SecureAccount("Pranav", 5000)
# print(acc.__balance) # Error! Direct access blocked.

acc.balance = 7000     # Calls setter safely
print(acc.balance)     # Calls getter -> Output: 7000
acc.balance = -100     # Output: Error: Balance negative nahi ho sakta!
```

#### Write OOP Classes to handle the following scenarios :-
* A user can create and view 2D Coordinates 
* A user can find out distance btw 2 coordinates 
* A user can find the distance of coordinate from origin 
* A user can check if a point lies on a given line 
* A user can find the distance btw a given 2D point and a given line  

````python
#Classes regarding coordinate geometry


class Point:
    def __init__(self,x,y):
        self.x_coord=x
        self.y_coord=y

    def __str__(self):
        return f"({self.x_coord},{self.y_coord})"

    def origin_distance(self):
        return ((self.x_coord**2)+(self.y_coord**2))**0.5

    #Distance btw two coordinates is called euclidean distance 
    def euclidean_distance(self,other):
        x1=(other.x_coord-self.x_coord)**2
        y1=(other.y_coord-self.y_coord)**2
        return (x1+y1)**0.5


class Line:
    def __init__(self,A=0,B=0,C=0):
        self.x_coeff=A
        self.y_coeff=B
        self.const_coeff=C

    def __str__(self):
        return  f"{(self.x_coeff)}x+{(self.y_coeff)}y+{(self.const_coeff)}" 

    def point_on_line(self,other):
        ans=self.x_coeff*other.x_coord+self.y_coeff*other.y_coord+self.const_coeff
        if ans==0:
            return "Yes"
        else:
            return "No"

    def distance_of_point_from_line(self,other):
        num=abs(self.x_coeff*other.x_coord+self.y_coeff*other.y_coord+self.const_coeff)
        den=((self.x_coeff**2)+(self.y_coeff**2))**0.5
        return num/den    
        

point1=Point(2,6)
point2=Point(6,13**0.5)
print(point1)
print(f"Distance of {point1} from origin {point1.origin_distance()}")  
print(f"Distance of {point1} from origin {point2.origin_distance()}") 

print(f"Distance btw {point1} and {point2} is  {point1.euclidean_distance(point2)}")
line1=Line(2,-1,1)
print(line1)

print(f"Is point {point1} lies on line {line1}?:{line1.point_on_line(point1)}")

print(f"Distance of point {point1} from line {line1} is {line1.distance_of_point_from_line(point1)}")
````

# ATM Banking App — OOP Concepts + Static Variable (Customer ID) 🏧

Yeh file tumhare ATM class code ko base bana ke, step-by-step samjhati hai ki **Unique Customer ID** wala feature kaise implement karna hai, aur is process mein **Instance Variable vs Static (Class) Variable** ka fark bhi clear kar degi.

---

## Step 1: Pehle chhota sa recap — tumhara current code kya karta hai

```python
class Atm:
    def __init__(self):
        self.pin = ""
        self.balance = 0
        self.Menu()
```

- `self.pin` aur `self.balance` — yeh **instance variables** hain.
- Instance variable ka matlab: **har object (customer) ke paas apni ALAG copy hoti hai** is variable ki.
- Jab tum `c1 = Atm()` banate ho, toh Python RAM mein ek naya "box" banata hai jisme `c1.pin`, `c1.balance` store hote hain — bilkul alag from `c2.pin`, `c2.balance`.

Yeh sahi hai kyunki **pin aur balance har customer ka apna-apna hona chahiye.** ✅

---

## Step 2: Manager ki new requirement — Unique Customer ID

Manager bola: *"Har customer ka ek unique CID (Customer ID) hona chahiye."*

Tumne try kiya:

```python
class Atm:
    def __init__(self):
        self.cid = 0
        self.cid += 1        # 0 + 1 = 1
        self.pin = ""
        self.balance = 0
```

Phir 3 objects banaye:

```python
c1 = Atm()
c2 = Atm()
c3 = Atm()

print(c1.cid)   # 1
print(c2.cid)   # 1
print(c3.cid)   # 1
```

**Result:** Teeno ka CID `1` aa raha hai. ❌ Yeh galat hai — expected tha `1, 2, 3`.

---

## Step 3: Yeh problem kyun aa rahi hai? (Root Cause)

Yaad karo — `self.cid` ek **instance variable** hai.

Jab bhi tum `Atm()` call karte ho:
1. Python ek **naya, fresh object** banata hai.
2. Us naye object ke andar `self.cid = 0` chalta hai → **CID reset ho jata hai 0 pe.**
3. Fir `self.cid += 1` chalta hai → CID ban jata hai `1`.
4. Yeh sequence **har naye object ke liye repeat hoti hai**, kyunki har object ka `self.cid` **apni memory mein alag se store hota hai.**

> 💡 **Simple analogy:** Socho har customer ko ek **apna khud ka notepad** diya gaya hai jisme woh CID likhta hai. Har naya customer apne khud ke blank notepad pe "0" likh ke phir "+1" karke "1" likh deta hai. Sabka notepad alag hai, isliye sabka answer "1" hi aata hai — kisi ko pata nahi ki dusre ka number kya tha!

**Conclusion:** Ek **shared, common counter** chahiye jo saare objects ke beech "connected" ho — taki jab ek customer bane, toh counter update ho aur **agla customer usi updated counter ko dekhe.** Isके liye chahiye — **Static Variable (Class Variable).**

---

## Step 4: Static Variable (Class Variable) kya hota hai?

| | **Instance Variable** | **Static / Class Variable** |
|---|---|---|
| **Kahan likhte hain** | `__init__` (constructor) ke andar, `self.` ke saath | Class ke andar, **constructor ke bahar**, seedha class body mein |
| **Memory** | Har object ki apni alag copy | **Ek hi copy**, saare objects ke beech **shared** |
| **Access karte kaise hain** | `self.variable_name` | `ClassName.variable_name` (ya `self.variable_name` — sirf *read* karne ke liye) |
| **Use kab karna hai** | Jab data har object ka **alag-alag** hona chahiye (pin, balance) | Jab data **sabme common/shared** hona chahiye (counter, bank name, interest rate) |
| **Real-life analogy** | Har customer ka apna bank passbook | Bank ka ek shared "token counter" jo reception pe lagा hai — sabke liye same |

**Syntax example:**

```python
class Atm:
    counter = 0        # 👈 STATIC VARIABLE — class ke andar, __init__ ke bahar

    def __init__(self):
        self.pin = ""          # instance variable
        self.balance = 0       # instance variable
```

Yahan `counter` **sirf ek baar** memory mein banta hai — jab tum class define karte ho, object banane se pehle hi. Saare objects isi ek `counter` ko **share** karenge.

---

## Step 5: Static Variable use karke Unique CID implement karna

```python
class Atm:
    counter = 0    # STATIC VARIABLE — sabhi objects ke beech shared

    def __init__(self):
        Atm.counter += 1        # ⚠️ IMPORTANT: ClassName ke through update karo
        self.cid = Atm.counter  # is customer ka permanent, unique CID
        self.pin = ""
        self.balance = 0
```

Ab check karte hain:

```python
c1 = Atm()
c2 = Atm()
c3 = Atm()

print(c1.cid)   # 1
print(c2.cid)   # 2
print(c3.cid)   # 3
```

Ab teeno customers ke alag-alag, sequential CID mil rahe hain! ✅

---

## Step 6: ⚠️ Sabse Bada Gotcha (Common Mistake)

Tum sochoge — `Atm.counter += 1` ki jagah `self.counter += 1` kyun nahi likh sakte?

```python
# ❌ GALAT TARIKA
def __init__(self):
    self.counter += 1     # Yeh naya PROBLEM create karega
    self.cid = self.counter
```

**Kya hota hai is se:**
- `self.counter += 1` Python mein internally iska matlab hai: `self.counter = self.counter + 1`
- Python pehle `self.counter` ko **read** karega — chunki instance ke paas apna `counter` nahi hai, toh woh **class variable ko dhundhega aur uski value use karega (read-only)**.
- Lekin jaise hi tum `=` (assignment) karte ho, Python **ek NAYA instance variable `self.counter` bana deta hai** us object ke liye — jo ab class variable `counter` ko **shadow (chhupa)** kar deta hai!
- Result: Har object ka apna alag `self.counter` ban jayega, aur **static variable wala poora concept fail ho jayega** — wahi purani problem wapas aa jayegi.

> 🔑 **Golden Rule:** Static variable ko **modify (update)** karne ke liye hamesha `ClassName.variable_name` use karo (jaise `Atm.counter += 1`). `self.variable_name` sirf **padhne (read)** ke liye safe hai, **likhne (write/update)** ke liye nahi.

---

## Quick Recap (TL;DR)

1. **Instance variable** (`self.x`) → har object ki apni copy → data jo **object-specific** hai (pin, balance).
2. **Static/Class variable** (`ClassName.x` ya class body mein bina `self` ke) → **ek hi shared copy** saare objects ke liye → data jo **common/global** hai (counter, bank interest rate, total customers).
3. Static variable **update** karne ke liye hamesha `ClassName.variable += 1` likho — `self.variable += 1` mat likho, warna naya shadow instance variable ban jayega.
4. Static variable ko **read** karne ke liye `self.variable` bhi chalega (kyunki Python object ke andar na milne pe class mein dhundh leta hai), lekin **safe practice** hai `ClassName.variable` hi consistently use karna.

# Encapsulation ki Kahani - Part 5: The Static Sabotage & The Ultimate Security 🛑

## Part 1: Static Variable Par Attack (Junior Ka Naya Khel)

Junior Developer ko company se nikaal diya gaya tha, lekin nikalne se pehle usne ek aakhiri shararat ki. Usne dekha ki ATM class ke andar ek **Static Variable (Class Variable)** hai jo bank ka naam store karta hai: `bank_name = "Global Bank"`. 

Junior ne socha, *"Agar main object ke balance ko nahi badal pa raha hoon, toh main poori bank ka naam hi badal kar fraud company likh dunga! Static variable toh bina object ke bhi badla ja sakta hai."*

Usne class ke bahar bina kisi object reference ke, seedhe Class Name ka use karke static variable ko corrupt kar diya:
```python
Atm.bank_name = "Chor Bank Ltd."  # Direct attack on Class Variable!
```
**Nateeja:** Jaise hi bank ka naam badla, saare customers ke ATM screens par galat naam dikhne laga aur afra-tafri mach gayi.

---

## Part 2: Senior Ka Masterstroke - Class-Level Getters & Setters

Senior Developer ne turant is attack ko block karne ke liye dimaag lagaya. Usne kaha, *"Ab main static variable ko bhi private (`__bank_name`) banaunga. Aur ise control karne ke liye main normal getter-setter nahi, balki **Class methods** ka use karke Class-Level Getters and Setters banaunga!"*

### Core Concept: Class Name vs Instance Reference 🧠
* **Normal Getters/Setters:** `obj.balance` se call hote hain kyunki woh object ke variables hain.
* **Static Getters/Setters:** `Atm.get_bank_name()` ya `Atm.set_bank_name()` se call hote hain. Inka kisi object (`self`) se lena-dena nahi hota, yeh seedhe **Class Name** ke zariye pure class level par validation check karte hain.

## Summary of the Entire Story 🧠
1. **Instance Variables (`self.var`)** ko secure karne ke liye normal functions/properties banate hain jo `obj.var` se chalti hain.
2. **Static Variables (`Class.var`)** ko secure karne ke liye `@classmethod` lagakar controllers banate hain, jo bina object ke seedhe **`ClassName.method()`** se access hote hain. Is tarah se poora system harami juniors se 100% bulletproof ho jata hai!
