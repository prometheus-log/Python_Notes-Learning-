## 1. Class Relationships: "Is-A" vs "Has-A" (Beginner)

Before diving deep into inheritance, it is critical to understand how classes relate to each other. There are two primary types of relationships:

1. **"Is-A" Relationship (Inheritance):** A subclass is a specialized version of the parent class. 
   * *Example:* A `Dog` **is a** `Animal`. A `Car` **is a** `Vehicle`.
2. **"Has-A" Relationship (Association/Composition):** A class contains an instance of another class as one of its attributes.
   * *Example:* A `Car` **has an** `Engine`. A `University` **has a** `Department`.

---

## 2. Introduction to Inheritance (Beginner)

Inheritance allows a new class (Child/Sub class) to adopt the attributes and methods of an existing class (Parent/Base class). This promotes code reusability.

### Syntax
```python
class Parent:
    # Parent features
    pass

class Child(Parent):
    # Child inherits Parent features + adds its own
    pass
```

### Basic Example ("Is-A" Relationship)
```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."

# Dog inherits from Animal
class Dog(Animal):
    def bark(self):
        return f"{self.name} says Woof!"

# Usage
my_dog = Dog("Buddy")
print(my_dog.eat())  # Inherited method: Buddy is eating.
print(my_dog.bark()) # Child method: Buddy says Woof!
```

## 3. Deep Dive into "Has-A" Relationships (Intermediate)

When classes interact without inheriting from one another, they rely on **Association**. This is divided into **Aggregation** and **Composition** based on how tightly bound the objects are.

### A. Aggregation (Weak "Has-A")
In Aggregation, the child object **can exist independently** of the parent object. If the parent is destroyed, the child survives.

```python
class Professor:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, name):
        self.name = name
        self.professors = []

    def add_professor(self, prof):
        self.professors.append(prof)

# Usage
prof1 = Professor("Dr. Smith")
math_dept = Department("Mathematics")
math_dept.add_professor(prof1)

# If we delete the department, the professor still exists
del math_dept
print(prof1.name)  # Output: Dr. Smith (Survived!)
```

### B. Composition (Strong "Has-A")
In Composition, the child object **cannot exist independently** of the parent object. The parent owns the lifetime of the child. If the parent is deleted, the child is automatically deleted.

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

class Car:
    def __init__(self, make, hp):
        self.make = make
        # The Engine object is created inside the Car constructor
        self.engine = Engine(hp) 

# Usage
my_car = Car("Toyota", 150)
print(my_car.engine.horsepower)  # Output: 150

# If we delete the car, the engine is destroyed along with it
del my_car
# print(my_car.engine) -> Raises NameError because engine is gone.
```

## 4. Types of Inheritance (Intermediate)

Python supports five distinct types of inheritance models.

### A. Single Inheritance
A single child class inherits from a single parent class.
```python
class Parent:
    def feature1(self): return "Feature 1"

class Child(Parent):
    def feature2(self): return "Feature 2"
```

### B. Multiple Inheritance
A child class can inherit from more than one parent class.
```python
class Father:
    def skills(self): return "Gardening"

class Mother:
    def talents(self): return "Painting"

class Child(Father, Mother):
    pass

c = Child()
print(c.skills())  # Gardening
print(c.talents()) # Painting
```

### C. Multilevel Inheritance
A child class inherits from a parent class, which in turn inherits from another parent class (Grandparent -> Parent -> Child).
```python
class Grandparent:
    def gold(self): return "10kg Gold"

class Parent(Grandparent):
    def land(self): return "2 Acres Land"

class Child(Parent):
    def car(self): return "Sports Car"
```

### D. Hierarchical Inheritance
Multiple child classes inherit from a single parent class.
```python
class Vehicle:
    def info(self): return "This is a vehicle"

class Car(Vehicle): pass
class Bike(Vehicle): pass
```

### E. Hybrid Inheritance
A blend of more than one type of inheritance (e.g., combining Multiple and Multilevel).

## 5. The `super()` Function & Method Overriding (Intermediate)

### Method Overriding
If a child class defines a method with the same name as a method in the parent class, the child's method overrides the parent's method.

### Using `super()`
The `super()` function allows you to call methods of the parent class inside the child class, which is vital when overriding `__init__`.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def get_details(self):
        return f"Name: {self.name}, Salary: {self.salary}"

class Manager(Employee):
    def __init__(self, name, salary, department):
        # Call the parent constructor to initialize name and salary
        super().__init__(name, salary)
        self.department = department

    # Overriding get_details
    def get_details(self):
        # Call parent's get_details and extend it
        base_details = super().get_details()
        return f"{base_details}, Department: {self.department}"

mgr = Manager("Alice", 90000, "IT")
print(mgr.get_details()) 
# Output: Name: Alice, Salary: 90000, Department: IT
```

## 6. Method Resolution Order - MRO (Advanced)

When dealing with multiple inheritance, Python needs a way to decide which class's method to run if the same method exists in multiple parents. This is called the **Method Resolution Order (MRO)**.

Python uses the **C3 Linearization algorithm** to construct the MRO.

### The Diamond Problem Example
```python
    A
   / \
  B   C
   \ /
    D
```

```python
class A:
    def show(self): print("In A")

class B(A):
    def show(self): print("In B")

class C(A):
    def show(self): print("In C")

class D(B, C):
    pass

obj = D()
obj.show() # Output: In B

# To view the exact order Python follows:
print(D.__mro__)
# Output: (<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>)
```

## 7. Abstract Base Classes - ABC (Advanced)

An **Abstract Class** acts as a blueprint for other classes. It allows you to create a set of methods that *must* be created within any child classes built from the abstract class. 

* Python doesn't provide abstract classes by default. We use the `abc` module.
* You cannot instantiate an abstract class.

```python
from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass

    @abstractmethod
    def refund(self, transaction_id):
        pass

class PayPal(PaymentGateway):
    def process_payment(self, amount):
        return f"Processing ${amount} via PayPal."
    
    def refund(self, transaction_id):
        return f"Refunding transaction {transaction_id} via PayPal."

# If you don't implement both methods in PayPal, Python will throw a TypeError.
gateway = PayPal()
print(gateway.process_payment(100))
```

## 8. Subclass Customization with `__init_subclass__` (Expert)

Introduced in Python 3.6, `__init_subclass__` allows a parent class to automatically detect, configure, or validate any class that inherits from it at the time of creation (before instantiation).

```python
class PluginBase:
    subclasses = []

    def __init_subclass__(cls, **kwargs):
        super().__init_subclass__(**kwargs)
        # Automatically register any subclass that inherits from PluginBase
        cls.subclasses.append(cls)

class AudioPlugin(PluginBase):
    pass

class VideoPlugin(PluginBase):
    pass

print(PluginBase.subclasses)
# Output: [<class '__main__.AudioPlugin'>, <class '__main__.VideoPlugin'>]
```

# Python OOP: Method Overriding (Beginner to Advanced)

**Method Overriding** tab hota hai jab ek Child class (Subclass) apne Parent class ke kisi method ko fir se apne tarike se define karti hai. 

Iska use tab kiya jata hai jab parent class ka method child class ke kaam ka na ho, ya child class ko usme kuch **extra changes** ya **features** add karne hon.

---

## 1. Method Overriding ke Important Rules
1. **Inheritance Zaroori Hai:** Bina parent-child relationship ke overriding nahi ho sakti.
2. **Method Name Same Hona Chahiye:** Jo naam parent class ke method ka hai, exact wahi naam child class ke method ka bhi hona chahiye.
3. **Arguments/Parameters Same Hone Chahiye:** Method ke andar pass hone wale variables ka signature match hona chahiye.

---

## 2. Basic Example (Complete Behavior Change)
Is example mein, child class parent ke method ko poori tarah se badal (replace) degi.

```python
class User:
    def home_screen(self):
        print("Welcome to the Standard User Dashboard.")

class Student(User):
    # Parent ke home_screen method ko override kar diya
    def home_screen(self):
        print("Welcome to the Student Dashboard! Apne enrolled courses dekho.")

# Usage
base_user = User()
student = Student()

base_user.home_screen()  # Output: Welcome to the Standard User Dashboard.
student.home_screen()    # Output: Welcome to the Student Dashboard! Apne enrolled courses dekho.
```

## 3. `super()` Function ka Use (Parent Behavior ko Extend Karna)
Real life mein, hum aksar parent ke logic ko fekte nahi hain. Hum chahte hain ki **parent wala code bhi chale, aur uske baad child ka apna extra code bhi chale**. Iske liye hum `super()` ka use karte hain.

### Constructor Overriding (`__init__`)
Ye sabsay zyada use hota hai. Agar aap child class mein `__init__` bana rahe ho, toh parent ke variables ko active rakhne ke liye `super().__init__()` call karna hi padega.

```python
class User:
    def __init__(self, username):
        self.username = username

    def display_profile(self):
        print(f"User: {self.username}")

class Student(User):
    def __init__(self, username, roll_no):
        # 1. Parent ke constructor ko call kiya taaki username set ho jaye
        super().__init__(username) 
        # 2. Child ka apna naya attribute add kiya
        self.roll_no = roll_no

    # Method override karke uski power badhai
    def display_profile(self):
        # Parent ka normal display chalega
        super().display_profile() 
        # Saath mein child ka extra data bhi print hoga
        print(f"Roll Number: {self.roll_no}")

# Usage
student1 = Student("Samarth", 101)
student1.display_profile()
# Output:
# User: Samarth
# Roll Number: 101
```

## 4. Advanced Concept: Polymorphism mein Use (Real-World Case)
Method overriding asal mein **Runtime Polymorphism** ki jaan hai. Isse aap ek hi loop ya function se alag-alag classes ke overridden methods ko dynamically run kar sakte ho.

```python
class Notification:
    def send(self):
        # Agar kisi child ne ise override nahi kiya toh ye error dega
        raise NotImplementedError("Subclass must implement this method")

class EmailNotification(Notification):
    def send(self):
        print("Sending SMTP Email securely...")

class SMSNotification(Notification):
    def send(self):
        print("Sending SMS via Twilio Gateway...")

# Ye function nahi janta kaunsa object aayega, ye bas .send() chala dega
def dispatch_alert(notification_object):
    notification_object.send()

# Usage
dispatch_alert(EmailNotification()) # Output: Sending SMTP Email securely...
dispatch_alert(SMSNotification())   # Output: Sending SMS via Twilio Gateway...
```

#### Inheritance Summary:-

- Constructor, attributes, methods get inherited to the child class

- The parent has no access to the child class

- Private properties of parent are not accessible directly in child class

- Child class can override the attributes or methods. This is called method overriding

- super() is an inbuilt function which is used to invoke the parent class methods and constructor

____

#### super() Summary:-

- Parent Class Access: super() ka main kaam child class ke andar se parent class ke methods aur variables (attributes) ko access karna hota hai. 

- Constructor Chaining: Agar child class mein __init__ constructor hai, toh parent class ke attributes ko initialize karne ke liye sabsay pehle super(). __init__() call karna mandatory hota hai.

- Code Reusability (Don't Repeat Yourself): Iska use tab kiya jata hai jab aapko parent class ka existing logic poori tarah badalna nahi hai, balki uske code ko barkarar rakhte hue apna extra code add (extend) karna hai.

- MRO (Method Resolution Order) Resolution: Multiple inheritance ke case mein, super() manually kisi class ka naam liye bina Python ke C3 Linearization (MRO) algorithm ke mutabik sahi order mein next parent class ko dhoodh kar call karta hai.

- No self Argument Needed: Jab aap super().method_name() likhte hain, toh aapko argument mein self pass karne ki zaroorat nahi hoti; Python ise background mein automatic handle kar leta hai.

- Attributes Access: super() se aap parent class ke methods ko access kar sakte ho, lekin direct instance attributes (jaise self.name) ko access nahi kar sakte. Unhe use karne ke liye parent ke constructor ya method ko hi call karna padega.

- Parameters Pass Karna: Haan, super().__init__(param1, param2) ki tarah aap parent class ke method ke mutabik parameters bilkul pass kar sakte ho. 

- Class Ke Bahar Use: Nahi, super() ko aap class ke bahar (global scope mein) use nahi kar sakte. Ise hamesha child class ke methods ke andar hi likha jata hai.

- Self Pass Nahi Karna: super().method() chalate waqt bracket ke andar self pass nahi kiya jata, Python ise khud handle karta hai.

### 💡 Yeh Kaise Kaam Kiya? (MRO Rule)
Python pehle Left-to-Right search karta hai. Chunki humne `class TeachingAssistant(Faculty, Student)` likha tha, isliye Python ne pehle **Faculty** ko priority di aur wahan method milte hi ruk gaya.
___
**MRO (Method Resolution Order)** wo order (rasta) hai jise Python follow karta hai kisi method ya attribute ko search karne ke liye, khaaskar jab aap **Multiple ya Hybrid Inheritance** use kar rahe hon.

Python iske liye **C3 Linearization algorithm** ka use karta hai taaki koi ambiguity na ho.

---

### Method Resolution Order:-

#### 1. MRO ke 3 Golden Rules (Shortcut Tricks)

Aapko lamba algorithm lagane ki zaroorat nahi hai, bas ye teen rules yaad rakho:
1. **Subclass hamesha Parent se pehle aayegi:** Agar `class B(A)` hai, toh MRO mein pehle `B` search hoga, fir `A`.
2. **Left-to-Right Priority:** Agar `class C(B, A)` hai, toh pehle Left wala (`B`) search hoga, fir Right wala (`A`).
3. **No Duplication:** Kisi bhi class ko MRO chain mein do baar nahi likha ja sakta. Wo apne sabsay aakhiri priority wale spot par rahegi.

# C3 Linearization Algorithm Formula & Notation

 ## C3 Linearization woh core algorithm hai jise Python (Python 2.3 ke baad se) use karta hai apni MRO (Method Resolution Order) list taiyar karne ke liye.
 - Jab aap code mein Multiple ya Hybrid Inheritance use karte hain, toh background mein yahi mathematical algorithm decide karta hai ki kaunsi class pehle search hogi aur kaunsi baad mein.
-  Isme "C3" ka matlab hai teen main properties jo ye guarantee karta hai:
    - Consistent Local Precedence Order: Class definition mein aapne jis order mein parents likhe hain (Left-to-Right), woh maintain rahega.

    -  Monotonicity: Agar Class A pehle aati hai Class B se kisi parent ke order mein, toh wahi sequence poori sub-class chain mein barkarar rahega.

    - Valid Extended Precedence: Ye aapas mein takraane wale rasto (conflicts) ko hal karta hai.

Python me MRO (Method Resolution Order) calculate karne ke liye background me **C3 Linearization Algorithm** chalta hai. Agar aapko iska mathematical representation kisi Markdown (`.md`) file me chahiye, toh iska proper LaTeX code aur explanation neeche diya gaya hai:

---

### 1. Main Formula

Kisi bhi Class $C$ ka MRO (jiske parents $P_1, P_2, \dots, P_N$ hain), is formula se nikala jata hai:

$$L[C(P_1, P_2, \dots, P_N)] = [C] + \text{merge}(L[P_1], L[P_2], \dots, L[P_N], [P_1, P_2, \dots, P_N])$$

---

### 2. Formula Components Explained

* **$L[C]$**: This represents the final linearized MRO list of class $C$.
* **$[C]$**: Ek single element ki list jisme wahi class hoti hai jiska MRO nikal rahe hain.
* **$+$**: Yeh list concatenation (jodhne) ka sign hai.
* **$\text{merge}(\dots)$**: Yeh woh primary function hai jo alag-alag parent lists ko aapas me combine karta hai.

---

### 3. Head-Tail Notation (The Merge Rule)

Merge function ko perform karne ke liye lists ko **Head** aur **Tail** me toda jata hai:

$$\text{If } \text{List} = [X, Y, Z] \implies \text{Head} = X, \quad \text{Tail} = [Y, Z]$$

#### The Selection Criteria:
Aap pehli list ka $\text{Head}$ (let's say $H$) uthate hain. Agar yeh $H$ baaki bachi hui kisi bhi list ke $\text{Tail}$ me **maujood nahi hai**, toh yeh selection safe mana jata hai:

$$\text{Next Element} = H \iff H \notin \text{Tail of any remaining list}$$

# Python OOP: Polymorphism (Beginner to Advanced)

**Polymorphic** ka simple matlab hota hai: **"Many Forms" (Ek cheez, alag-alag roop)**. 
Object-Oriented Programming (OOP) mein jab ek hi method name ya operator alag-alag data types ya classes ke liye alag tarike se behave karta hai, toh use **Polymorphism** kehte hain.

> **Note:** Python mein traditional *Method Overloading* (same name, different arguments) natively support nahi hoti, isliye is guide mein hum purely Pythonic Polymorphism ke real concepts (Overriding, Duck Typing, Operator Overloading) par focus karenge.

---

## 1. Polymorphism Ki Branches (Types)

Python mein Polymorphism ko mukhhya roop se teen branches mein banta jata hai:

1. **Compile-Time / Static Polymorphism:** Python dynamic language hai, isliye isme strict compile-time polymorphism nahi hota. Lekin **Operator Overloading** ko is category mein dekha jata hai.
2. **Runtime / Dynamic Polymorphism:** Yeh **Method Overriding** ke zariye hota hai, jahan decide hota hai ki runtime par kis class ka method chalega.
3. **Duck Typing:** Yeh Python ka sabse khaas aur exclusive dynamic behavior hai.

---

## 2. Basics: Operators Aur Built-in Functions Mein Polymorphism (Beginner)

OOPs classes par jaane se pehle, aap rozana Python mein polymorphism use karte ho:

### A. Operator Polymorphism (`+` Operator)
`+` sign numbers ko jodta hai, lekin strings ko aapas mein chipka (concatenate) deta hai.

```python
# Numbers ke saath (Addition)
print(5 + 10)       # Output: 15

# Strings ke saath (Concatenation)
print("Samarth" + " " + "Python")  # Output: Samarth Python

# Lists ke saath (Merge)
print([1, 2] +)  # Output: [1, 2, 3, 4]
```

### B. Function Polymorphism (`len()` Function)
`len()` function ek hi hai, lekin ye list, string, ya dictionary sabke andar ka data alag tarike se ginta hai.

```python
print(len("Samarth"))       # Output: 7 (Characters count kiye)
print(len([10, 20, 30]))     # Output: 3 (Elements count kiye)
```

## 3. Duck Typing in Python (Intermediate)

Python ka ek basic rule hai: **"If it walks like a duck and quacks like a duck, it’s a duck."**
Iska matlab, Python yeh nahi dekhta ki object kis class ka hai (type kya hai); Python bas yeh dekhta hai ki us object ke paas woh required **method** maujood hai ya nahi.

### Practical Code Example:
Maan lo hamare paas do alag classes hain jinme koi inheritance ka rishta nahi hai, par dono ke paas same naam ka method hai.

```python
class VSCode:
    def execute_code(self):
        print("Compiling code, running syntax checks, executing output...")

class Notepad:
    def execute_code(self):
        print("Saving text file, running basic script via terminal...")

# Yeh function kisi class se bndha nahi hai. 
# Isko bas ek aisa object chahiye jiske paas 'execute_code' method ho.
def run_my_program(editor_object):
    editor_object.execute_code()  # Polymorphic call

# Usage
run_my_program(VSCode())   # Output: Compiling code, running syntax checks...
run_my_program(Notepad())  # Output: Saving text file, running basic script...
```

## 4. Runtime Polymorphism via Method Overriding (Advanced)

Yeh tab hota hai jab ek Parent-Child relationship (Inheritance) ho, aur child class parent ke method ko badal de. Ek single interface ya loop se hum alag-alag objects ko bina unka type jaane handle kar sakte hain.

```python
class PaymentGateway:
    def process_payment(self, amount):
        # Base implementation (Ya toh empty chodh do ya default logic rakho)
        pass

class UPIPayment(PaymentGateway):
    def process_payment(self, amount):
        print(f"Processing ₹{amount} securely via UPI (Scanning QR Code)...")

class CreditCardPayment(PaymentGateway):
    def process_payment(self, amount):
        print(f"Processing ₹{amount} via Credit Card Gateway (Validating OTP)...")

# --- Polymorphism in Action (The Common Interface) ---
# Yeh function nahi jaanta ki user UPI se pay karega ya Card se.
# Yeh bas dynamic_object ke andar ka process_payment chala deta hai.
def checkout(payment_object, bill_amount):
    payment_object.process_payment(bill_amount)

# Usage
order1 = UPIPayment()
order2 = CreditCardPayment()

checkout(order1, 500)   # Output: Processing ₹500 securely via UPI...
checkout(order2, 2500)  # Output: Processing ₹2500 via Credit Card...
```

## 5. Operator Overloading: Custom Objects Mein Polymorphism (Expert)

Agar aap apni khud ki banayi hui classes ke objects ko `+`, `-`, `<`, ya `>` operators ke saath chalana chahte hain, toh aapko Python ke **Magic Methods** (Dunder Methods) ko override karna padta hai. Ise **Operator Overloading** kehte hain.

### Real-World Example: Financial Wallets Ko Plus Karna
Maan lo aapke paas do wallet ke objects hain, aur aap chahte ho ki `wallet1 + wallet2` karne par unke amounts aapas mein jud jayein.

```python
class Wallet:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    # '+' operator ko is class ke liye define karne ka magic method
    def __add__(self, other):
        # self = pehla wallet, other = dusra wallet
        total_balance = self.balance + other.balance
        return f"Combined Balance: ₹{total_balance}"

    # Print karne par kaisa dikhega uske liye magic method
    def __str__(self):
        return f"{self.owner}'s Wallet (₹{self.balance})"

# Usage
w1 = Wallet("Samarth", 1500)
w2 = Wallet("Pushpendra", 2500)

print(w1)  # Output: Samarth's Wallet (₹1500)
print(w2)  # Output: Pushpendra's Wallet (₹2500)

# Polymorphism! '+' ab Wallet ke objects ko bhi jod sakta hai
result = w1 + w2
print(result)  # Output: Combined Balance: ₹4000
```
