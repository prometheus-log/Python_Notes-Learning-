````python
#Generators:- An easier way of creating iterators but way to harder to understand 

#Creating my own range function 

class myownrange:
    def __init__(self,start,end):
        self.start=start
        self.end=end


    def __iter__(self):
        return iterator_of_my_range(self)

class iterator_of_my_range:
    def __init__(self,Iterable_object):
        self.iterator=Iterable_object
    def __iter__(self):
        return self
    def __next__(self):
        if self.iterator.start>=self.iterator.end:
            raise StopIteration
        
        current=self.iterator.start
        self.iterator.start+=1
        return current

for i in myownrange(1,11):
    print(i)    


#Ab jaise iterators banane ke liye you first need a iterable object then in iterator object so everything feels a bit cluttered .

#Python devs ne socha kyu na isko thoda decluttered format ya easier tareeke se banane ki koshish kare to generators were born !!!!            
````

# Benefits of Generators and Generator Chaining

## Benefits of Generators
* **Memory Efficiency**: Generates elements one at a time on-the-fly, keeping RAM consumption extremely low regardless of data size.
* **Lazy Evaluation**: Pauses execution and computes data only when requested, saving execution time and processing power.
* **Infinite Streams**: Handles endless data sources (like live server logs or continuous sensor data) safely without crashing.
* **Cleaner Syntax**: Achieves full iterator behavior using a simple function and the `yield` keyword, replacing complex dunder class code.

## Generator Chaining (As seen in the image)
Generator chaining is a technique where the output of one generator is passed directly as the input into another generator. This builds a pipeline where data flows through different processing steps one item at a time without creating intermediate lists in memory.

### Code Breakdown:
```python
def fibonacci_numbers(nums):
    x, y = 0, 1
    for _ in range(nums):
        x, y = y, x + y
        yield x

def square(nums):
    for num in nums:
        yield num ** 2

print(sum(square(fibonacci_numbers(10))))
```
