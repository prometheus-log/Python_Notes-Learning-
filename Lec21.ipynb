```python
#Creation of dataframe requires a 2D list if creating from list
#The list which is created below usme har row mainn is order se data bhara hai iq, marks,package_secured 

# https://drive.google.com/drive/folder... is link se dataset download karle aage jaake use honge (ony movues and ipl_matches hi use honge)
import numpy as np
import pandas as pd
student_data=[
    [100,80,10],
    [90,70,7],
    [120,100,14],
    [80,50,2]
    
]
pd.dataframe(student_data,columns=["iq","marks","package"])
#as you can see columns parameter main list of columns to be created pass karoge 
```

```python
#Creating from a dictionary 
student_dict = {
    'name':['nitish','ankit','rupesh','rishabh','amit','ankita'],
    'iq':[100,90,120,80,0,0],
    'marks':[80,70,100,50,0,0],
    'package':[10,7,14,2,0,0]
}
#as you can see har key se ek list of data  associated hai since keys to column ka naam ban hi jayengi to unke  adar ki series ya data bhi dena hi hoga 

students = pd.DataFrame(student_dict) #yahan you can see since keys column ban jayegi to columns parameter dena nhi pada 
print(students)

students.set_index('name',inplace=True) #diresct student dataframe main change karega and jo  internally index allocate hue hai uski jagah name ko index bana dega  row based 
#inplace=True se direct original dataframe main change  hoga gar False hota ya nhi hota too alag dataframe banata usme changes karke wo return karta 
students
```

```python
#Filtering a data frame 
matches

#jitne ipl hue hai har saal ke winner ka naam batao (find all winners )(season se pata chal jayega ki saal kounsa hai winning team se winner pata chal  jayega and then har saal ka winner chahiye yni final jeetne wala yni match nhumber se final  mil jayega)

#Sabse pehle ye nikalunga ki kounse kounse matches are final ?
mask=matches["MatchNumber"]=="Final"

new_df=matches[mask]
#ab mere pass ek aisi dataframe hai jiska har maych final hai ab inke winners chhaiy e and season

new_df[["Season","WinningTeam"]]
```

```python
#Out of all ipl matches kitne matches main super over hua hai ?
matches

#ab mereko woh saare matches chahiye jimne super over hua tha yni matches ka column but super over wala 

mask=matches["SuperOver"]=="Y"
new_df=matches[mask]
#ab mujhe ye jaana hai ki kitne matches hue hai therefore mujhe number of rows batana hai 
new_df.shape[0]
```

```python
#csk ne kolkata  main kitne match jite ?
matches
mask=matches["WinningTeam"]=="Chennai Super Kings"
new_Df=matches[mask]
mask2=new_Df["City"]=="Kolkata"
new_df2=new_Df[mask2]
new_df2.shape[0]

#isko aur easily bhi kar skte the bitwise and se 
improved_mask=((matches["City"]=="Kolkata") & (matches["WinningTeam"]=="Chennai Super Kings"))
new_df3=matches[improved_mask].shape[0]
print(new_df3)
```

```python
#toss winner is match winner in percentage (matlab toss jitne par kitne baar aise hua ki toss winning team is match winning team)
matches
mask=((matches["TossWinner"])==(matches["WinningTeam"]) )
new_df=matches[mask] #isme wo saari teams hai jisne toss jitne par match jita ab mujhe nikalna hai ki iska percenntage kya hai agar nikalnna hota ki kitne baar hua hai 

#now to calculate percentage 
percentage=(new_df.shape[0]/matches.shape[0])*100
print(percentage)
```
