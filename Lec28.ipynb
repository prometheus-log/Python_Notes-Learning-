```python
# plotting a simple function
price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]
#Samjho ek column is price column and other is price column .
#Here year is categorical  column since year ki category hai fixed while price is fluctuating numerical data 

#While plotting keep categorical column on x and numerical column on y (standard)
plt.plot(year,price)
```

```python
#Using a pandas dataframe 
batsman=pd.read_csv(r"C:\Users\Asus\Downloads\sharma-kohli.csv")
batsman #data abour rohit sharma and virat kohli and the total runs they made iin a season 

plt.plot(batsman["index"],batsman["V Kohli"]) #V kohli ki throughout all ipl ki performance ka plot 
```

```python
#Plotting multiple  plots:- ek saath ek hi graph pe multiple plotting 

#Suppose kisine bola ek hi graph main kohli and sharma ka side by side compairision  of track record 

plt.plot(batsman["index"],batsman["V Kohli"])
plt.plot(batsman["index"],batsman["RG Sharma"])
```

```python
#Ab graph to plot hogaya but dhnag se koi info nhi gayi koi newbie n=bina code padhe dekhega to usko thodi pata x axis pe kya value y axis pe kya chalo ek baar ko x axis as year samajh bhi lega to y  axis ka kya therefore we need labels.

#How to add labels /title to a graph
plt.plot(batsman["index"],batsman["V Kohli"])
plt.plot(batsman["index"],batsman["RG Sharma"])
plt.title("Rohit v/s Kohli performance comparirision throughout their career ")

plt.xlabel("Year")
plt.ylabel("Runs throughout career ")
```

```python
#Ab aisa hi kyu hua ki rohit ka curve is orange and kohli ka is blue what if i wanna change it ?

plt.plot(batsman["index"],batsman["V Kohli"],color="cyan")
plt.plot(batsman["index"],batsman["RG Sharma"],color="green")

#YAhan color parameter main hex codes bhi de skte ho if css padha hai then you know else its a code for colors in hexadecimal number system nhi aaya samajh google karle 

```

```python
#What if plot main line solid nhi chahiye kuch differennt chhaiye "linestyle" parameter ka use karrlo

#Default value is solid 
plt.plot(batsman["index"],batsman["V Kohli"],linestyle="dashed")
plt.plot(batsman["index"],batsman["RG Sharma"],linestyle="dashed")

#Different values  of line style you can try:-
#dotted,dashdot,
```

```python
#Lines ka width bhi badha skte ho 
plt.plot(batsman["index"],batsman["V Kohli"],linestyle="dashed",linewidth=3)
plt.plot(batsman["index"],batsman["RG Sharma"],linestyle="dotted",linewidth=2) 
```

```python
#Abhi jaise ye  sirf plot kar de raha hai par actually 2d plot uses points and then connect them in order to form a graph/plot 

#These points are called markers in matplotlib and we can manipulate them /see them 
plt.plot(batsman["index"],batsman["V Kohli"],linestyle="dashed",marker="d")
plt.plot(batsman["index"],batsman["RG Sharma"],linestyle="dotted",marker="d")

#Saare points diamond shape ke dikhenge and values which can be used for marker parameter 
# '.': Point, 'o': Circle, 's': Square'D':, Diamond (large),'d': Thin diamond, 'p': Pentagon , 'h': Hexagon (variant 1), 'H': Hexagon (variant 2), '8': Octagon

# 2. Triangles & Pointers
# '^': Triangle up
# 'v': Triangle down
# '<': Triangle left
# '>': Triangle right

# 3. Lines, Crosses, & Pluses
# '+': Plus sign
# 'x': X cross
# '*': Star
# '|': Vertical line segment
# '_': Horizontal line segment

#markersize se marker ka size bhi control kar skte ho takes numeric values 
```

```python
#Ab phir newbie ko thodi pata chal raha ki kounnsi color ki line is for kohli and sharma ?

#Labelling individually 
plt.plot(batsman["index"],batsman["V Kohli"],linestyle="dashed",marker="d",label="Virat")
plt.plot(batsman["index"],batsman["RG Sharma"],linestyle="dotted",marker="d",label="Rohit")

#Abhi bhi screen pe nhi dikhaya coz we haven't called legend function 
plt.legend()
#Ab left par hi legend kyu aaya  plot main ?

# loc parameter ke karan jiski default value is "best" jo ki best position dhund leta hai 
plt.legend(loc="upper right")
```

```python
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.xlabel("Year")
plt.ylabel("Price")

#Ab hoga kya ki koi ek value ke dusre se bahut bade hone ke karan graph flat dikhne laga and then ek sudden jump aaya variation dikhne khatam hogaye to ye problem aayegi (is sabse badi ya dusro se exceptionally badi value ko outlier bolte hai)

#We can trim the x axis and y axis to show only a particular amout of data
```

```python
# ab in dots ka size se main represent kar skta hoon ki kitna effective tha example dekho samjh jaoge 
tips=sns.load_dataset("tips")
tips

plt.scatter(tips["total_bill"],tips["tip"])
#Tips and total bill ka scatter plot mil jayega but ek baat aur yahan as you can see ki bill zyada hai to tip bhi zyada hogi n=and one of the reaon bill zyada ho skta hai  is ki no of customers zyada bill wale ke zyada hoo jaise ki ek bade group ka dine in 
```

```python
#SO kya main is plot ke dots ko aise represent kar skta hoon ki jiske customer zyada uski dot ka size bhi in accordance ho ?
plt.scatter(tips["total_bill"],tips["tip"],s=tips["size"]*20)
#Ab dekhoge ki ki jab 20 multiply nhi hoga na to size pf dots kaafi chota aayega kyunki jiska bill zyada us part main 2 ya 3 hi aise bande hote hai to isliye 20 se multiply kiya taaaki ache se represent ho sake 
```

```python
#Plotting scatter plot by using plt.plot() and not using plt.scatter 

plt.plot(tips["total_bill"],tips["tip"],"o")

#Only  problem is size wala nhi kar paoge and color badalne mainn bhi dikkkat hogi 
#Used because its much fater than plt.scatter 
```

```python
#How to plot a simple bar chart 
children=[10,20,40,10,30]
colors=["red","blue","green","yellow","pink "]

plt.bar(colors,children)
```

```python
plt.bar(colors,children,color="black")
```

```python
#2 types of bar charts exists:- vertical and horizontal upar wala vertical hai now horizontal banana hai to :-

plt.barh(colors,children)

#Better too use horizontal when no of categories zyada ho (>8)(kyunki verticla main max 8 tak clean dikh skta hai par uske baad cluttering hone lagti hai)
plt.grid(color="black")
```

```python
#Kayi baar requirement ke hisaab m=se multiple quantities need to be plotted on the same graph


season_record=pd.read_csv(r"C:\Users\Asus\Downloads\batsman_season_record.csv")
season_record
```

```python
#Bar chart banao saaare batsman ka ki unhone 2015 main kitne run banaye ?
plt.bar(season_record["batsman"],season_record["2015"])
```

```python
#Ab is upar wale graph par 2016 ka data bhi plot karna ho to kar to dunga par differentiate kaise hoga?
# kyunki bar ka rang bhale hi badal jaye par overlap karega to chup skta hai na !

#Example:- 
plt.bar(season_record["batsman"],season_record["2015"])
plt.bar(season_record["batsman"],season_record["2016"])
```

```python
#Better ye  hota ki dono graph side by  side print hote buildings ki tarah it would look clean plus differentitate kar paoge 

#THoda jugaad lagana padta hai mehnat karni padti hai 
#Pehle width ghatao then 2015 ke har data ko  thoda left khiskao
plt.bar(np.arange(season_record.shape[0])-0.2,season_record["2015"],width=0.2)
plt.bar(np.arange(season_record.shape[0]),season_record["2016"],width=0.2)
plt.bar(np.arange(season_record.shape[0])+0.2,season_record["2017"],width=0.2)
# plt.bar(season_record.shape[0])
#Ab jaise dekhoge to perfect teeno years ka compairision to dikhh raha par bar shift karne ke lliye x axis se name hata diye in order toshift but how to get them back ?

plt.xticks(np.arange(season_record.shape[0]), season_record["batsman"])

```

```python
#A new kind of problem :-
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
#Ab isme kya hoga jaise categories ka naam kaafi bada hai and x axis pe dikhne ki jagah dusri categories ke naam ke saath overlap karega 
```

```python
#How to solve by using xticks:-
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children)
plt.xticks(rotation='vertical') #Rotation parameter 
```

```python
#Stacked bar chart :- multiple plotting hui par sab aapas main jud gaya 

#Suppose saare batman ne 2015,2016,2017 main total milakar ek batsman ne kitne run banaye aisa saare batman ke liye nikalna hai but in this process mujhe yeh bhi retain karna hai ki 2015,16,17 main individually kitne banaye ?

plt.bar(season_record["batsman"],season_record["2015"],label="2015")
plt.bar(season_record["batsman"],season_record["2016"],bottom=season_record["2015"],label="2016")
total_15_16 = season_record["2015"] + season_record["2016"]
plt.bar(season_record["batsman"],season_record["2017"],bottom=total_15_16,label="2017")
```
