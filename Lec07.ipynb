# Python Object-Oriented Programming (OOP): Classes and Objects

### What is Object-Oriented Programming (OOP)?Object-Oriented Programming (OOP) ek programming paradigm (programming karne ka tareeka) hai jo poore software design ko "Functions and Logic" ke bajaye "Objects and Data" ke aaspas organize karta hai.

## Part 1: Beginner Level - The Blueprint and The Instance

### 1. What is a Class and an Object?
* **Class:** Yeh ek user-defined blueprint ya template hota hai jiska use karke hum objects banate hain. Yeh attributes (data) aur methods (functions) ko ek single unit mein bind karta hai.
* **Object:** Yeh class ka ek real-world instance (copy) hota hai. Jab tak sirf class banti hai, memory allocate nahi hoti. Memory tabhi milti hai jab class ka object banta hai.

### 2. Constructor (`__init__`) and `self`
* **`__init__` Method:** Ise constructor bolte hain. Jab bhi class ka koi naya object banta hai, yeh method background mein automatically run ho jata hai. Iska main kaam object ke variables ko initialize karna hai.
* **`self` Parameter:** Yeh current object ko represent karta hai. Iska use karke hum class ke andar uske variables aur functions ko access karte hain.

### Beginner Example
```python
class Smartphone:
    # Constructor
    def __init__(self, brand, model, price):
        self.brand = brand      # Instance Attribute
        self.model = model      # Instance Attribute
        self.price = price      # Instance Attribute

    # Instance Method
    def display_info(self):
        return f"Phone: {self.brand} {self.model} costs ${self.price}"

# Creating Objects (Instances)
phone1 = Smartphone("Apple", "iPhone 15", 999)
phone2 = Smartphone("Samsung", "Galaxy S24", 899)

print(phone1.display_info())  # Output: Phone: Apple iPhone 15 costs $999
print(phone2.display_info())  # Output: Phone: Samsung Galaxy S24 costs $899
```

## Part 2: Intermediate Level - Attributes, Methods, and Data Hiding

### 1. Instance vs Class Attributes & Methods
* **Instance Members:** Yeh har ek object ke liye alag aur unique hote hain. Inhe `self` ke through access kiya jata hai.
* **Class Members:** Yeh poori class ke liye common hote hain. Agar class attribute ko change karoge, toh woh saare objects ke liye change ho jayega. Inhe `@classmethod` decorator aur `cls` parameter ke sath banate hain.
* **Static Methods:** Inka na toh object (`self`) se lena-dena hota hai aur na hi class (`cls`) se. Yeh normal utility functions hote hain jo bas organization ke liye class ke andar rakh diye jate hain. Inhe `@staticmethod` se likhte hain.

### 2. Data Hiding (Encapsulation)
Python mein data ko secure rakhne ke liye access modifiers ka use hota hai:
* **Public:** Normal variable name (e.g., `self.brand`) - Ise kahin se bhi access kiya ja sakta hai.
* **Protected:** Single underscore (e.g., `self._premium_user`) - Yeh bas ek warning signal hai ki ise class ke bahar use mat karo.
* **Private:** Double underscore (e.g., `self.__bank_balance`) - Ise class ke bahar directly access nahi kiya ja sakta (Python iska naam change kar deta hai jise Name Mangling bolte hain).

### Intermediate Example
```python
class BankAccount:
    bank_name = "Global Bank"  # Class Attribute

    def __init__(self, owner, balance):
        self.owner = owner          # Public Attribute
        self.__balance = balance    # Private Attribute (Encapsulation)

    # Class Method
    @classmethod
    def change_bank_name(cls, new_name):
        cls.bank_name = new_name

    # Static Method
    @staticmethod
    def is_valid_interest_rate(rate):
        return 0 <= rate <= 15

    # Getter Method (Private data ko safely access karne ke liye)
    def get_balance(self):
        return self.__balance

# Usage
account = BankAccount("Rahul", 5000)
print(account.owner)          # Output: Rahul
# print(account.__balance)   # Error! Direct access nahi milega.

print(account.get_balance())  # Output: 5000 (Safely accessed via getter)
```

## Part 3: Advanced Level - Magic Methods & Operator Overloading

Python mein **Magic Methods** (ya **Dunder Methods**, kyunki inke naam ke aage-piche double under score `__` hota hai) aapke custom objects ko built-in functions aur operators ke sath behave karne ki super-power dete hain.

Jab hum in methods ko apni class mein overwrite karte hain taaki operators (+, -, *, /) humare objects ke sath kaam kar sakein, toh ise **Operator Overloading** kehte hain.

### Core Dunder Methods Explained:
* **`__str__(self)`**: Jab aap object ko `print()` karte ho ya `str()` mein convert karte ho, toh screen par kya user-friendly text dikhna chahiye, yeh yeh method taiyar karta hai.
* **`__add__(self, other)`**: Yeh `+` operator ke behavior ko control karta hai.
* **`__sub__(self, other)`**: Yeh `-` operator ke behavior ko control karta hai.
* **`__mul__(self, other)`**: Yeh `*` operator ke behavior ko control karta hai.
* **`__truediv__(self, other)`**: Yeh `/` (division) operator ke behavior ko control karta hai.

### Advanced Example (Operator Overloading with a Vector2D Class)
Chalo dekhte hain ki do mathematical vectors (co-ordinates) ko aapas mein plus, minus ya multiply kaise karte hain:

```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # 1. Custom String Representation (Object print hone par kya dikhega)
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    # 2. Overloading '+' Operator
    def __add__(self, other):
        if isinstance(other, Vector2D):
            return Vector2D(self.x + other.x, self.y + other.y)
        return NotImplemented

    # 3. Overloading '-' Operator
    def __sub__(self, other):
        if isinstance(other, Vector2D):
            return Vector2D(self.x - other.x, self.y - other.y)
        return NotImplemented

    # 4. Overloading '*' Operator (Scalar Multiplication)
    def __mul__(self, scalar):
        if isinstance(scalar, (int, float)):
            return Vector2D(self.x * scalar, self.y * scalar)
        return NotImplemented

    # 5. Overloading '/' Operator
    def __truediv__(self, scalar):
        if isinstance(scalar, (int, float)):
            if scalar == 0:
                raise ZeroDivisionError("Vector ko zero se divide nahi kar sakte!")
            return Vector2D(self.x / scalar, self.y / scalar)
        return NotImplemented

# --- Testing the Advanced Code ---

v1 = Vector2D(6, 8)
v2 = Vector2D(2, 4)

# __str__ check karo
print(v1)          # Output: Vector(6, 8)

# __add__ check karo (+)
v_add = v1 + v2
print(v_add)       # Output: Vector(8, 12)

# __sub__ check karo (-)
v_sub = v1 - v2
print(v_sub)       # Output: Vector(4, 4)

# __mul__ check karo (*)
v_mul = v1 * 3
print(v_mul)       # Output: Vector(18, 24)

# __truediv__ check karo (/)
v_div = v1 / 2
print(v_div)       # Output: Vector(3.0, 4.0)
```

# Python Magic Methods (Dunder Methods) & Their Superpowers

## Introduction: What are Dunder Methods?
Python mein **Magic Methods** un special methods ko kaha jata hai jinke naam ke aage aur piche double underscore (`__`) hota hai. Isiliye inhe **Dunder (Double Under)** methods bhi bolte hain. 

### The Core Superpower 🦸‍♂️
Inki sabse badi superpower yeh hai ki **inhe aapko explicitly (manually) call nahi karna padta**. Jab bhi aap Python ke kisi built-in operator (jaise `+`, `-`, `*`) ya functions (jaise `print()`, `len()`) ka use apne custom object par karte ho, toh Python background mein in dunder methods ko automatically trigger kar deta है। 

Ise OOPs ki bhasha mein **Operator Overloading** aur **Polymorphism** kehte hain.

Yeh methods har class ki buniyad hote hain aur object ke banne, print hone aur khatam hone ko manage karte hain.

* **Benefits:** `Sabse bada fayda constructor call karne ka yeh hai ki isme hum wah sari cheezein de skte hai jiska control hum user ko nhi dena chahte

## Part 2: Intermediate Level - Mathematical Operator Overloading

Agar aap chahte hain ki aapke banaye gaye do custom objects aapas mein mathematical operators (`+`, `-`, `*`, `/`) ke sath kaam karein, toh aapko in methods ko override karna padega.

| Operator | Magic Method | Trigger Condition |
| :--- | :--- | :--- |
| `+` | `__add__(self, other)` | `obj1 + obj2` |
| `-` | `__sub__(self, other)` | `obj1 - obj2` |
| `*` | `__mul__(self, other)` | `obj1 * obj2` |
| `/` | `__truediv__(self, other)` | `obj1 / obj2` |

### Intermediate Code Example (E-Commerce Cart System):
Maan lijiye humare paas do shopping carts hain aur hum `+` operator use karke dono ka total bill jodna chahte hain:

```python
class Cart:
    def __init__(self, items_count, total_price):
        self.items_count = items_count
        self.total_price = total_price

    def __str__(self):
        return f"Cart has {self.items_count} items. Total: ₹{self.total_price}"

    # Overloading the '+' operator
    def __add__(self, other):
        if isinstance(other, Cart):
            # Ek naya Cart object return kar rahe hain dono ko jodkar
            new_items = self.items_count + other.items_count
            new_price = self.total_price + other.total_price
            return Cart(new_items, new_price)
        return NotImplemented

# Execution
cart1 = Cart(3, 1500)
cart2 = Cart(2, 800)

final_cart = cart1 + cart2  # Triggers __add__ background mein
print(final_cart)           # Output: Cart has 5 items. Total: ₹2300
```

## Part 3: Advanced Level - Built-in Function Behaviors & Context Managers

Advanced programming mein dunder methods ka use karke hum apne custom objects ko bilkul Python ke built-in data types (jaise list ya dictionary) jaisa shaktishali bana sakte hain.

### 1. `__len__(self)`
* **Superpower:** Aapke object ke paas kitna data hai, uski length nikalne ke liye use hota hai.
* **Automatic Trigger:** `len(obj)`

### 2. `__eq__(self, other)`
* **Superpower:** Default roop se Python do objects ke memory address ko compare karta hai. `__eq__` ko use karke hum yeh tay kar sakte hain ki do objects ke andar ka data barabar (`==`) hai ya nahi.
* **Automatic Trigger:** `obj1 == obj2`

### 3. `__call__(self, ...)`
* **Superpower:** Yeh ek bohot hi shaandar method hai. Iski madad se aap **apne object ko ek normal function ki tarah call** kar sakte hain parenthesis `()` lagakar.
* **Automatic Trigger:** `obj()`

### Advanced Code Example:
```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    # 1. len() control karne ke liye
    def __len__(self):
        return self.pages

    # 2. '==' operator control karne ke liye
    def __eq__(self, other):
        if isinstance(other, Book):
            return self.title == other.title and self.pages == other.pages
        return False

    # 3. Object ko call karne ke liye (Superpower)
    def __call__(self, reader_name):
        print(f"{reader_name} is currently reading '{self.title}'...")


# --- Verification and Testing ---

book1 = Book("Python Mastery", 450)
book2 = Book("Python Mastery", 450)
book3 = Book("Data Structures", 600)

# Testing __len__
print(len(book1))        # Output: 450

# Testing __eq__ (==)
print(book1 == book2)    # Output: True (kyunki content same hai)
print(book1 == book3)    # Output: False

# Testing __call__ (Object behaves like a function)
book1("Rahul")          # Output: Rahul is currently reading 'Python Mastery'...
```

## Summary Cheat Sheet

| Category | Dunder Method | Real-world Trigger | What it overrides |
| :--- | :--- | :--- | :--- |
| **Lifecycle** | `__init__` | `Obj()` | Object Creation / Constructor |
| **String Output** | `__str__` | `print(obj)` | Readable string format |
| **Math** | `__add__`, `__sub__` | `+`, `-` | Arithmetic operations |
| **Comparison** | `__eq__` | `==` | Equality logic |
| **Built-in** | `__len__` | `len(obj)` | Size measurement |
| **Execution** | `__call__` | `obj()` | Function-like invocation |


````python
#Designing our own object class known as Fractions
class Fractions:
    #Parametrized constructor
    def __init__(self,x,y):
        self.num=x
        self.den=y  #ye self. den islioye karte hain kyunki yeh python ko batata  hai is variable ko object ki memeory main fix kardo yeh local variable nhi hai jo class khtm hote hi the end hojayega !

    def __str__ (self):
        return f"{self.num}/{self.den}"
    #ye batayega ki fraction dikhega kaisa ?
    #iska superpower yeh hai ki jab bhi hum apne object ko print() main daalenge toh python object ko print karne ke liye samajhna chahega ki object kaise hain to woh directly iss dunder method main ghusega  and figure out karega kya print karna hai 
     

    #Fractions addition logic 
    def __add__(self,other):
        print("idhar",self,other)

        lcm=self.den *other.den
        add= ((lcm/self.den) *self.num) + ((lcm/other.den) *other.num)
        #add=self.num*other.den + other.num*self.den ye bhi karte to aajata 
        return f"{add}/{lcm}"


    #automatically triggers jab aap apni class ke 2 objects beech main + operator ka istemal karoge.Also it requires two parameters self and other and is case main jo pehla argument pass karoge na wo self acquire karega  similarly other dusre ko acquire karega dono ke pass apna num and den hoga  

    #Fraction Subtraction logic
    def __sub__(self,other):
        lcm=self.den *other.den
        sub= ((lcm/self.den) *self.num) - ((lcm/other.den) *other.num)
        #add=self.num*other.den + other.num*self.den ye bhi karte to aajata 
        return f"{sub}/{lcm}"

    #working same as __add__ dunder method but  for subtraction

    #Multiplication logic
    def __mul__(self,other):
        return f"{self.num*other.num}/{self.den*other.den}"

    #Division logic
    def __truediv__(self,other):
        return f"{self.num*other.den}/{self.den*other.num}"

    #Decimal conversion
    def convert_to_decimal(self):
        return self.num/self.den 


fr1=Fractions(3,4)
fr2=Fractions(5,6)

print(fr1,fr2)
print(fr1+fr2)
print(fr1-fr2)
print(fr1*fr2)
print(fr1/fr2)
````
