# `np.histogram()` — Complete Guide

## 1. Kya hai ye?

`np.histogram()` data ko **bins (buckets)** mein daal ke count karta hai ki har bucket mein kitni values aayi. Ye **sirf numbers calculate karta hai** — koi graph nahi banata. Graph banane ke liye matplotlib chahiye (neeche cover kiya hai).

## 2. Syntax

```python
np.histogram(a, bins=10, range=None)
```

| Parameter | Matlab |
|---|---|
| `a` | Tera data array |
| `bins` | Kitne buckets banane hain (default 10) |
| `range` | `(min, max)` tuple — na do to data ke actual min-max se le lega |

## 3. Working — step by step

**Step 1: Range decide karta hai**
Data ka minimum aur maximum dekhta hai (ya jo `range` mein diya). Ye poora range bins mein divide hoga.

**Step 2: Range ko equal-width bins mein todta hai**
Range `[0, 10]` aur `bins=5` ho to har bin ki width = `(10-0)/5 = 2`.
Bins: `[0,2), [2,4), [4,6), [6,8), [8,10]`

**Step 3: Har data point ko check karta hai kis bin mein fit hota hai**

**Step 4: Us bin ka counter +1 karta hai**

**Step 5: Do arrays return karta hai**
- `hist` → counts array (length = `bins`)
- `bin_edges` → boundaries array (length = `bins + 1`)

## 4. Bin inclusivity rule (important!)

Har bin `[a, b)` hoti hai — **left included, right excluded**.

- `a` (left edge) → agar value bilkul `a` ke barabar hai, is bin mein aayegi
- `b` (right edge) → agar value bilkul `b` ke barabar hai, is bin mein **nahi** aayegi, agle bin mein jaayegi

**Exception:** sirf **last bin** dono taraf inclusive hoti hai `[a, b]` — kyunki uske aage koi bin hi nahi hai.

### Example

```python
import numpy as np

data = [0, 2, 4, 6, 8, 10]
hist, bin_edges = np.histogram(data, bins=5, range=(0, 10))

print(bin_edges)  # [0. 2. 4. 6. 8. 10.]
print(hist)        # [1 1 1 1 2]
```

Bins:
1. `[0, 2)` → sirf `0` → count = 1
2. `[2, 4)` → sirf `2` → count = 1
3. `[4, 6)` → sirf `4` → count = 1
4. `[6, 8)` → sirf `6` → count = 1
5. `[8, 10]` ← **last bin, dono taraf inclusive** → `8` aur `10` dono → count = 2

> **Yaad rakhne ka trick:** Bins ek train ke coaches hain. Har coach ka entry gate front pe hai (left included), exit gate peeche band hai (right excluded) — jo front se ghusega wahi uska passenger. Sirf **last coach ka peeche wala gate bhi khula hai** (train khatam ho rahi hai, koi agla coach nahi).

## 5. Full example

```python
import numpy as np

data = [1, 2, 2, 3, 5, 7, 8, 9]
hist, bin_edges = np.histogram(data, bins=4)

print(hist)        # [3 1 1 3]
print(bin_edges)   # [1. 3. 5. 7. 9.]
```

- `[1,3)` → 1,2,2 → 3
- `[3,5)` → 3 → 1
- `[5,7)` → 5 → 1
- `[7,9]` → 7,8,9 → 3 (last bin, dono inclusive)

## 6. Matplotlib ka istemal karke histogram dikhana

### Method A — Direct (easiest, recommended)

`plt.hist()` internally `np.histogram` jaisa hi logic use karta hai counting ke liye, phir seedha bar chart bana deta hai.

```python
import matplotlib.pyplot as plt

data = [1, 2, 2, 3, 5, 7, 8, 9]

plt.hist(data, bins=4, edgecolor='black')
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

Bas itna hi — `plt.hist()` khud `np.histogram` wala kaam kar leta hai andar hi andar.

### Method B — Manual (jab tumhe `np.histogram` ke output pe control chahiye)

Kabhi kabhi tumhe pehle `hist` aur `bin_edges` alag se calculate karne hote hain (jaise kisi aur calculation ke liye use karna ho), aur phir sirf plot karna ho. Tab `plt.bar()` use karte hain:

```python
import numpy as np
import matplotlib.pyplot as plt

data = [1, 2, 2, 3, 5, 7, 8, 9]

# Step 1: histogram calculate karo
hist, bin_edges = np.histogram(data, bins=4)

# Step 2: bar chart banao using bin edges
plt.bar(bin_edges[:-1], hist, width=np.diff(bin_edges), edgecolor='black', align='edge')
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram (manual using np.histogram)")
plt.show()
```

**Kyun `bin_edges[:-1]`?** Kyunki `bin_edges` mein `bins+1` values hoti hain (edges), lekin bars sirf `bins` hi honge. Isliye last edge chhod dete hain — har bar apne **left edge** se start hoti hai.

**Kyun `width=np.diff(bin_edges)`?** Taaki har bar ki width uske actual bin-width ke barabar ho (agar bins equal-width nahi hain to ye zaroori hai).

### Kab Method A vs Method B use karein?

| Situation | Method |
|---|---|
| Sirf visualize karna hai | Method A (`plt.hist`) |
| Counts pe aage calculation bhi karni hai (e.g., normalize karna, compare karna) | Method B (`np.histogram` + `plt.bar`) |

## 7. Quick summary

- `np.histogram()` → sirf counting karta hai, koi plot nahi
- Bins `[a, b)` hoti hain, sirf last bin `[a, b]`
- `plt.hist()` → direct plotting, khud counting kar leta hai
- `np.histogram()` + `plt.bar()` → jab counts pe control chahiye ho pehle
