```python
#Sequence sum:-
#1/1!+2/2!+3/3!----------------till n terms
n=int(input("Enter your number"))
sum=0
for i in range(1,n+1,1):
    x=i
    for j in range(x-1,1,-1):
        x*=j
    sum+=i/x
print("Sum of following sequence=",sum)
```

```python
#PAttern:-

#1
#121
#12321
#1234321

n=int(input("Enter number of rows:"))

for i in range(1,n+1):
    x=1
    for j in range(1,2*i):
       if(j>i-1):
           print(x,end='')
           x-=1  
       else:
           print(x,end='')
           x+=1
    print()       
```

```python
#Slicing:Extracting multiple characters at once 

s="hello world"

print(s[0:5]) # 0 se 4 tak hi jayenge 

print(s[:4]) #s[:] dunga ti kuch bhi slice  nhi hoga 

# you can even provide a step size sorta thing

print(s[::2]) # n step size matlab n-1 ka jump

print(s[6:0:-1])

#Code for string reversal
print(s[::-1])

print(s[-5:])

print(s[-1:-6:-1])
```

```python
# Find length of string without using len() function 
name=input("Enter your desired string")
count=0
for i in name:
    count+=1
print("Length of string {} is {}".format(name,count))    
```

```python
#Extract username from the given email
# if mail is samarthpathrotkar.2006@gmail.com

mail=input("Enter desired mail")
username=""
for i in mail:
    if i=='@':
        break
    else:
        username+=i
print("username of {} is {}".format(mail,username))
```

```python
#Count frequency of a particular character in a provided string 

string=input("Enter string:")
substring=input("Enter substring for frequency :")
print("Frequency of substring {} in string {} is {}".format(substring,string,string.count(substring)))
```

```python
#Removing a particular character from string 

string=input("Enter string:")
remove=input("Enter substring for frequency :")
result=""

for i in string:
    if i ==remove:
        continue
    else:
        result+=i
print("String after particular removal is {}".format(result))        
```

```python
# Write a problem to check whether given string is palindrome or not 

string=input("Enter string for palindrome checking:")
n=len(string)
if string[0:]==string[-1:-(n+1):-1]:
    print("String is palindromic")
else:
    print("String ain't palindromic")  

```
