#  Tuples in Python

A **tuple** is a built-in, ordered, and immutable collection data type in Python(similar to list). Tuples are commonly used to store sequences of heterogeneous data (data of different types) and are safer than lists because their data cannot be altered after creation.

---

## 1. Key Characteristics of Tuples

*   **Ordered:** Elements have a defined order that will not change.
*   **Immutable:** Once created, elements cannot be added, removed, or modified.
*   **Allow Duplicates:** Can contain multiple elements with the same value.
*   **Heterogeneous:** Can store a mix of different data types (e.g., strings, integers, objects).

---

## 2. Creating Tuples

### Standard Initialization
Tuples are defined using parentheses `()` and separating elements with commas.

```python
# Empty tuple
empty_tuple = ()

# Tuple with multiple data types
person_tuple = ("Amit", 28, "Software Engineer")
```

### The Single-Element Trap
To create a tuple with only one element, you **must** include a trailing comma. Otherwise, Python treats it as a regular data type wrapped in parentheses.

```python
# Incorrect: This creates an integer
not_a_tuple = (5)  # type(not_a_tuple) -> int

# Correct: This creates a tuple
single_element_tuple = (5,)  # type(single_element_tuple) -> tuple
```

### Tuple Constructor
You can convert other iterables (like lists) into a tuple using the `tuple()` constructor.

```python
num_list = [1, 2, 3]
converted_tuple = tuple(num_list)
```

---

## 3. Accessing Tuple Elements

### Indexing and Slicing
Tuples support zero-based indexing and negative indexing, similar to strings and lists.

```python
coordinates = (10, 20, 30, 40)

# Positive Indexing
print(coordinates[0])   # Output: 10

# Negative Indexing
print(coordinates[-1])  # Output: 40

# Slicing (Start:Stop:Step)
print(coordinates[1:3]) # Output: (20, 30)
```

---

## 4. Immutability Explained

Attempting to change a tuple item will raise a `TypeError`.

```python
fruits = ("apple", "banana")
# fruits[0] = "cherry"  # Raises TypeError
```

### Workaround for Modification
If you absolutely need to modify a tuple, convert it to a list, change it, and convert it back.

```python
colors = ("red", "green", "blue")

# Convert to list
temp_list = list(colors)
temp_list[1] = "yellow"

# Convert back to tuple
colors = tuple(temp_list)
print(colors)  # Output: ('red', 'yellow', 'blue')
```

---

## 5. Tuple Operations and Methods

Since tuples are immutable, they support limited built-in methods compared to lists.

### Built-in Methods
Python provides only two built-in methods for tuples:

*   `count(value)`: Returns the number of times a value appears.
*   `index(value)`: Returns the index of the first occurrence of a value.

```python
mix = (1, 2, 3, 2, 4)

print(mix.count(2))  # Output: 2
print(mix.index(3))  # Output: 2
```

### Tuple Unpacking
Extracting tuple values directly into variables.

```python
point = (4, 5)
x, y = point

print(x)  # Output: 4
print(y)  # Output: 5
```

---

# Sets in Python: Beginner to Advanced

A **set** is an un-indexed, unordered, and mutable collection of unique elements in Python. Sets are based on a data structure called a **Hash Table**, making membership tests (checking if an item exists) highly efficient.

---

## 1. Key Characteristics of Sets

*   **Unordered & Un-indexed:** Elements do not have a defined order or position. You cannot access items using indexes (e.g., `my_set[0]` will raise a `TypeError`).
*   **Unique Elements:** Duplicate values are automatically eliminated.
*   **Mutable (Mostly):** You can add or remove items from a set, but the items within the set must be of an immutable (hashable) data type (like strings, integers, or tuples).
*   **Highly Efficient:** Checking if an element exists in a set takes O(1) constant time complexity on average.

---

## 2. Creating Sets (Beginner)

### Standard Initialization
Sets are defined using curly braces `{}` or the `set()` constructor.

```python
# A standard set with unique elements
fruits = {"apple", "banana", "cherry", "apple"}
print(fruits)  # Output: {'banana', 'cherry', 'apple'} (Duplicates removed)
```

### The Empty Set Trap
You **cannot** create an empty set using `{}` because Python interprets empty curly braces as an empty dictionary.

```python
# Incorrect: This creates a dictionary
not_a_set = {}  # type(not_a_set) -> dict

# Correct: Use the set constructor
empty_set = set()  # type(empty_set) -> set
```

---

## 3. Basic Modifications (Beginner)

```python
prime_numbers = {2, 3, 5}

# Adding elements
prime_numbers.add(7)

# Adding multiple elements
prime_numbers.update([11, 13, 17])

# Removing elements (Method 1: remove - raises error if item missing)
prime_numbers.remove(3) 

# Removing elements (Method 2: discard - safe, does not raise error)
prime_numbers.discard(100) # 100 is not in set, but no error raised
```

---

## 4. Core Mathematical Operations (Intermediate)

Sets in Python mimic mathematical set theory flawlessly.

```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
```

### Union
Combines elements from both sets (removes duplicates).
```python
print(set_a.union(set_b))        # Output: {1, 2, 3, 4, 5, 6}
print(set_a | set_b)             # Alternative Syntax (Bitwise OR)
```

### Intersection
Returns elements common to both sets.
```python
print(set_a.intersection(set_b)) # Output: {3, 4}
print(set_a & set_b)             # Alternative Syntax (Bitwise AND)
```

### Difference
Returns elements present in `set_a` but not in `set_b`.
```python
print(set_a.difference(set_b))   # Output: {1, 2}
print(set_a - set_b)             # Alternative Syntax (Minus)
```

### Symmetric Difference
Returns elements that are in either of the sets, but not in both.
```python
print(set_a.symmetric_difference(set_b)) # Output: {1, 2, 5, 6}
print(set_a ^ set_b)                     # Alternative Syntax (Bitwise XOR)
```

---

## 5. Subset and Superset Relations (Intermediate)

```python
x = {1, 2}
y = {1, 2, 3, 4}

print(x.issubset(y))    # True (All elements of x are in y)
print(y.issuperset(x))  # True (y contains all elements of x)
print(x.isdisjoint(y))  # False (They share common elements)
```

---

## 6. Advanced Concepts 

### Hashing and the Mutability Rule
A set can only contain elements that are **hashable**. An object is hashable if it has a hash value that never changes during its lifetime (i.e., it needs to be immutable).

```python
# Valid: Strings and integers are hashable
valid_set = {"text", 123, (1, 2)} 

# Invalid: A list is mutable and cannot be hashed
# invalid_set = {[1, 2], 3} # Raises TypeError: unhashable type: 'list'
```

### Time & Space Complexity Analysis
Because Python sets use open-addressing hash tables internally:

| Operation | Average Case Complexity | Worst Case Complexity |
| :--- | :--- | :--- |
| **Add / Insert** | O(1) | O(n) (Due to hash collisions) |
| **Remove / Delete**| O(1) | O(n) |
| **In / Containment**| O(1) | O(n) |
| **Union (\(A \mid B\))** | O(len(A) + len(B)) | O(len(A) + len(B)) |

*Senior Dev Tip:* Always look for opportunities to convert lists into sets if you are performing heavy lookups inside loops (`if item in list` takes O(n), whereas `if item in set` takes O(1)).

### Set Comprehension
Similar to list comprehensions, you can dynamically build sets with built-in logic.

```python
# Create a set of squares for even numbers only
even_squares = {x**2 for x in range(10) if x % 2 == 0}
print(even_squares) # Output: {0, 16, 64, 4, 36}
```

### Frozenset: The Truly Immutable Set
Standard sets are mutable, meaning they cannot be used as dictionary keys or inside another set. Python provides `frozenset` to solve this problem.

```python
immutable_set = frozenset([1, 2, 3])
# immutable_set.add(4) # Raises AttributeError

# Frozensets can be used as keys in dictionaries
my_dict = {immutable_set: "Valid Key"}
```

| Feature | Set | List | Tuple |

# Dictionaries in Python: Beginner to Advanced

A **dictionary** (dict) is an ordered (as of Python 3.7+), mutable collection of key-value pairs. Dictionaries are highly optimized in Python and function as associative arrays or hash maps.

---

## 1. Key Characteristics of Dictionaries

*   **Key-Value Pairs:** Data is stored as `key: value` mappings.
*   **Keys Must Be Unique & Hashable:** You cannot have duplicate keys. Keys must be of an immutable type (strings, numbers, or tuples).
*   **Values Can Be Anything:** Values can be duplicated and can hold any data type (lists, sets, other dicts).
*   **Ordered:** Python 3.7+ guarantees that dictionary items maintain their insertion order.
*   **Highly Efficient:** Lookup, insertion, and deletion operate in O(1) average time complexity.

---

## 2. Creating and Accessing Dictionaries (Beginner)

### Initialization
```python
# Standard literal initialization
user_profile = {
    "username": "coder_99",
    "email": "dev@example.com",
    "login_count": 5
}

# Using dict() constructor
config = dict(host="localhost", port=8080)
```

### Accessing Values Safely
Using square brackets `[]` will raise a `KeyError` if the key does not exist. The `.get()` method handles missing keys gracefully.

```python
# Risky: Raises KeyError if "status" is missing
# current_status = user_profile["status"] 

# Safe: Returns None if key is missing
current_status = user_profile.get("status")

# Safe with Fallback: Returns "Guest" if key is missing
role = user_profile.get("role", "Guest")
```

---

## 3. Modifying and Removing Items (Beginner)

```python
employee = {"name": "Rahul", "role": "Intern"}

# Adding or Updating
employee["role"] = "SDE-1"  # Updates existing key
employee["department"] = "Backend"  # Adds new key-value pair

# Removing using pop() (returns the removed value)
role_value = employee.pop("role")

# Removing the last inserted item (LIFO order)
last_item = employee.popitem()  # Returns tuple: (key, value)

# Deleting a key directly
del employee["name"]
```

---

## 4. Iterating and View Objects (Intermediate)

Dictionary views are dynamic. If the dictionary changes, the view updates automatically.

```python
payload = {"status": 200, "message": "Success", "data": []}

# Iterating over keys
for key in payload.keys():
    print(key)

# Iterating over values
for value in payload.values():
    print(value)

# Iterating over both (Unpacking items)
for key, value in payload.items():
    print(f"{key} -> {value}")
```

---

## 5. Advanced Concepts 

### Dictionary Comprehension
Dynamically transform or filter dictionaries.

```python
# Generate a dictionary of numbers and their cubes
cubes = {x: x**3 for x in range(1, 6)}
# Output: {1: 1, 2: 8, 3: 27, 4: 64, 5: 125}

# Filter dictionary based on values
scores = {"Alice": 85, "Bob": 40, "Charlie": 92}
passed_students = {k: v for k, v in scores.items() if v >= 50}
```

### Advanced Merging and Updating (Python 3.9+)
Use the merge operator (`|`) and update operator (`|=`) for cleaner dictionary manipulation.

```python
dict_a = {"x": 1, "y": 2}
dict_b = {"y": 99, "z": 4}

# Merge into a new dictionary (b overrides values of a on collision)
merged_dict = dict_a | dict_b  # {'x': 1, 'y': 99, 'z': 4}

# Update in-place
dict_a |= dict_b
```

### The Hash Table Architecture & Memory Optimization
Internally, Python dictionaries use a high-performance hash table design. 

1. **The Hash Function:** When you pass a key, Python computes `hash(key)` to determine its index location in an internal array.
2. **Compact Dicts (PEP 468):** Since Python 3.6, dictionaries use a tightly packed array for entries and a separate small sparse array for indices. This structural change reduced dictionary memory consumption by roughly **20% to 25%**.

```python
# Senior Dev Tip: Avoid mutating dictionaries while iterating over them. 
# It alters the internal hash table size and raises a RuntimeError.
# Instead, iterate over a copy: for k, v in dict.copy().items():
```

### Specialized Dictionary Variations
When native `dict` isn't enough, use the `collections` module:

*   **`defaultdict`:** Automatically initializes a default value if a requested key does not exist yet (eliminates explicit key checks).
*   **`Counter`:** A dictionary subclass designed explicitly for counting hashable objects.

```python
from collections import defaultdict, Counter

# Defaultdict example
word_counts = defaultdict(int)  # Default value for any new key is 0
word_counts["apple"] += 1

# Counter example
char_counts = Counter("abracadabra")
print(char_counts)  # Output: Counter({'a': 5, 'b': 2, 'r': 2, 'c': 1, 'd': 1})
```

## 6. Time Complexity Analysis

| Operation | Average Case | Worst Case | Notes |
| :--- | :--- | :--- | :--- |
| **Lookup / Get** | $O(1)$ | $O(n)$ | Worst case occurs only during heavy hash collisions. |

#  Dictionary Comprehension in Python: Beginner to Advanced

Dictionary comprehension is an elegant, concise, and highly optimized way to create or transform dictionaries from iterables in Python. It replaces bulky `for` loops with a single line of readable code.

---

## 1. Core Syntax & Architecture

The basic anatomy of a dictionary comprehension requires curly braces `{}` and a key-value mapping separated by a colon `:`:

```python
{key_expression: value_expression for item in iterable}
```

### The Standard Loop Equivalent
```python
# Traditional Loop Approach
squares = {}
for x in range(1, 4):
    squares[x] = x**2

# Dictionary Comprehension Equivalent
squares = {x: x**2 for x in range(1, 4)}
# Output: {1: 1, 2: 4, 3: 9}
```

---

## 2. Beginner Level: Basic Transformations

### Creating a Dictionary from Lists/Iterables
You can map elements or evaluate mathematical operations dynamically.

```python
# Transforming an existing list into a state dictionary
names = ["samarth", "rahul", "amit"]
user_lengths = {name: len(name) for name in names}
# Output: {'samarth': 7, 'rahul': 5, 'amit': 4}
```

### Using `zip()` to Merge Two Lists into a Dictionary
When you have separate lists for keys and values, pair them instantly.

```python
keys = ["name", "college", "semester"]
values = ["Samarth", "BGIEM", "5th"]

student_profile = {k: v for k, v in zip(keys, values)}
# Output: {'name': 'Samarth', 'college': 'BGIEM', 'semester': '5th'}
```

## 3. Intermediate Level: Conditional Logic

You can introduce filtering and inline updates using conditional statements (`if` and `if-else`).

### Filtering Data (`if` at the end)
The `if` condition filters out items from the source iterable *before* adding them to the dictionary.

```python
scores = {"Algorithms": 100, "Computation": 75, "DBMS": 45, "Cybersecurity": 97}

# Keep only subjects where marks are greater than 50
passed_subjects = {subject: marks for subject, marks in scores.items() if marks > 50}
# Output: {'Algorithms': 100, 'Computation': 75, 'Cybersecurity': 97}
```

### Transforming Values (`if-else` at the start)
Use `if-else` inside the value expression to conditionally modify the incoming data.

```python
# Mark status as 'Pass' or 'Fail' based on marks
grade_sheet = {sub: ("Pass" if marks >= 50 else "Fail") for sub, marks in scores.items()}
# Output: {'Algorithms': 'Pass', 'Computation': 'Pass', 'DBMS': 'Fail', 'Cybersecurity': 'Pass'}
```

## 4. Advanced Level: Senior Developer Engineering

### Handling Nested Dictionary Comprehensions
Nested comprehensions are highly powerful when working with multi-dimensional datasets, matrices, or relational mappings.

```python
# Production Data Matrix: Student marks across semesters
semester_data = {
    "Semester_1": {"Math": 85, "Physics": 90},
    "Semester_2": {"Math": 92, "Physics": 78}
}

# Scale up all marks by adding a 5-point grace buffer across all subjects
scaled_data = {
    sem: {subject: marks + 5 for subject, marks in subjects.items()} 
    for sem, subjects in semester_data.items()
}
# Output: {'Semester_1': {'Math': 90, 'Physics': 95}, 'Semester_2': {'Math': 97, 'Physics': 83}}
```

### Swapping Keys and Values (Inversion)
A standard data engineering pattern to invert unique dictionary mappings for reverse indexing.

```python
port_mappings = {"HTTP": 80, "HTTPS": 443, "SSH": 22}

# Invert keys and values
inverse_lookup = {port: protocol for protocol, port in port_mappings.items()}
# Output: {80: 'HTTP', 443: 'HTTPS', 22: 'SSH'}
```

### Performance & Memory Metrics
Dictionary comprehensions are not just syntactic sugar; they run faster than standard `for` loops.

1. **Bytecode Optimization:** Dictionary comprehension executes at C-level speed inside the Python interpreter using specialized `MAP_ADD` bytecode instructions. It eliminates the overhead of repeatedly looking up and calling the `.update()` or `__setitem__` methods.
2. **Readability vs Complexity:** While highly optimized, nesting beyond two levels degrades code maintainability.

```python
# Senior Dev Tip: If your dictionary comprehension stretches beyond 2 lines 
# or contains multiple complex nested conditions, break it down into a regular function 
# or loop to keep the codebase maintainable.
```
