```python
#Creating datetimeindex using python datetime 

pd.DatetimeIndex([dt.datetime(2026,9,12),dt.datetime(2026,9,13),dt.datetime(2026,9,14)])

#Creation using timestamps::-
dt_index=pd.DatetimeIndex([pd.Timestamp(2026,9,12),pd.Timestamp(2026,9,13),pd.Timestamp(2026,9,14)])
```

```python
#USing DateTimeIndex as series index 
pd.Series([1,2,3],index=dt_index)
```

```python
#date_range function generates dates in a give range 
pd.date_range(start="2026/9/12",end="2026/10/15",freq="D") #freq parameter ki default value is D only ad best part is its flexibility suppose 1 din chodh ke date chahiye  thi to fre=2D 2 din chodh ke freq=3D 
```

```python
#freq parameter main B diya to only business days print honge yni monday->friday sat sun will be ignored

#agar har hafte ka ek hi din chahiye  to W likhdo freq main and ye saare din  default sundays honge agar koi aur din chahiye  to W-THU aise likhdo ab Thu likha to thursdays aayenge 

pd.date_range(start="2026/9/12",end="2026/10/15",freq="W-THU")
```

```python
#Ghante  wise chahiye to freq="h" kardo ghante wise mil jayega minute wise ke liye freq=min  and second wise ke liye s 

pd.date_range(start="2026/9/12",end="2026/10/15",freq="h")
```

```python
#ME->month end wise ,MS->month start wise 
pd.date_range(start="2026/9/12",end="2026/10/15",freq="ME")

# Since given range main ek mahine end ho raha and ek start ho raha therefore ME and MS main ek hi date aayegi respectively 

#YE for  year end and YS for year start 
```

```python
#USing periods if  end date  you dont wanna specify 
pd.date_range(start="2026/9/12", periods=30,freq="D")
#Start se agle 30 din ka data 
```

```python
#to_datetime function 

# converting an existing object to pandas timestamp/datetimeindex object 

s=pd.Series(["2023/1/1","2022/1/1","2021/1/1"])
# ab inka dtype  object hai to date month year fetch  karne main string accessor ka use then split then functions bahut khichdi hai to isse better hai series pakad ke datetime object hi banado 

y=pd.to_datetime(s) #converted s to datetime object and stored inn y
y.dt.month_name()

#dt is datetime accessor 

#Now to_datetime() function main ek parameter hota hai  errors = "coercced" isse jaise suppose list min hai "2026/130/1" ab aise koi 130 mhina nhi hota to  coerced aisi date ko ignore karega bolega NaT  hai and then baaki pe kaam karke datetimeindex bana dega 
```
