# Part 1: Modules

## 1.1 Module kya hota hai

Jaise-jaise tumhara code bada hota hai, sab kuch **ek hi `.py` file** mein rakhna mushkil ho jaata hai — thousands of lines ek file mein scroll karna, dhoondhna, maintain karna, sab painful ho jaata hai.

**Module** simply ek `.py` file hoti hai jisme Python code (functions, classes, variables) hota hai, jise tum **kisi doosri file mein import karke reuse** kar sakte ho.

**Real life analogy:** Socho tumhare ghar mein alag-alag rooms hain — kitchen, bedroom, study room. Har room ka apna specific kaam hai. Agar sab kuch ek hi room mein daal do (bed, gas stove, books, sab), to dhoondhna mushkil ho jaayega. Modules bhi waisi hi separate "rooms" hain tumhare code ke liye — har module ek specific kaam ke liye responsible hota hai.

```mermaid
flowchart TD
    A["Bina Modules ke:\nsaara code ek 5000-line file mein"] --> A1["😵 Maintain karna mushkil\nDhoondhna mushkil\nReuse nahi ho sakta"]
    B["Modules ke saath:\nalag-alag .py files"] --> B1["math_utils.py"]
    B --> B2["string_utils.py"]
    B --> B3["main.py — sabko import karta hai"]

    style A1 fill:#ffcccc
```

## 1.2 Module Import Karna

Maan lo tumhare paas ek file hai `calculator.py`:

```python
# calculator.py
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

PI = 3.14159
```

Ab isko doosri file mein use karne ke **4 tareeke** hain:

### a) Poora module import karna

```python
import calculator

result = calculator.add(5, 3)     # module.function() ka tareeka
print(calculator.PI)
```

### b) Specific cheez import karna

```python
from calculator import add, PI

result = add(5, 3)      # ab direct naam se call kar sakte ho
print(PI)
```

### c) Sab kuch import karna (⚠️ generally avoid karo)

```python
from calculator import *

result = add(5, 3)      # sab kuch directly available hai
```

> **Warning:** `import *` avoid karo — isse pata nahi chalta ki koi function/variable kahaan se aaya. Agar do modules mein same naam ka function ho, to silently ek doosre ko **overwrite** kar dega, aur tumhe error bhi nahi aayega — bas galat result milega.

### d) Alias ke saath import karna

```python
import calculator as calc

result = calc.add(5, 3)     # chhota naam, likhne mein aasaan
```

```mermaid
flowchart LR
    A["import calculator"] --> A1["calculator.add(5,3)\nfull path se call"]
    B["from calculator import add"] --> B1["add(5,3)\ndirect naam se call"]
    C["import calculator as calc"] --> C1["calc.add(5,3)\nshort alias se call"]
```

## 1.3 Built-in vs User-defined Modules

| Type | Kya hai | Example |
|---|---|---|
| **Built-in Modules** | Python ke saath already installed, C mein likhe gaye, bahut fast | `sys`, `os`, `math`, `time` |
| **Standard Library Modules** | Python installation ke saath aate hain, Python mein hi likhe gaye | `json`, `csv`, `random`, `datetime` |
| **User-defined Modules** | Tumne khud banaye hain | `calculator.py`, `myutils.py` |
| **Third-party Modules** | `pip` se install karte ho, koi aur bana chuka hai | `requests`, `numpy`, `flask` |

```python
import math
print(math.sqrt(16))    # 4.0 — built-in module

import random
print(random.randint(1, 10))   # standard library module
```

## 1.4 `dir()` aur `help()`

Koi bhi module use karne se pehle uske andar kya-kya hai, ye jaanne ke liye do bahut useful functions hain:

## 1.5 `__name__ == "__main__"` ka raaz

Har Python file ka ek special variable hota hai `__name__`. Iski value depend karti hai ki file **kaise chalayi ja rahi hai**:

- Agar file **directly run** ki ja rahi hai (`python calculator.py`) → `__name__` ki value hoti hai `"__main__"`
- Agar file **kisi doosri file se import** ki ja rahi hai → `__name__` ki value hoti hai module ka naam (`"calculator"`)

```python
# calculator.py
def add(a, b):
    return a + b

print(f"Ye file ka naam hai: {__name__}")

if __name__ == "__main__":
    # Ye code SIRF tab chalega jab calculator.py directly run ho
    print("Testing add function:", add(2, 3))
```

Agar tum `python calculator.py` directly chalao → output: `Ye file ka naam hai: __main__` phir `Testing add function: 5`

Agar koi doosri file `import calculator` karti hai → sirf `Ye file ka naam hai: calculator` print hoga, `if` block ke andar wala code **nahi chalega**.

```mermaid
flowchart TD
    Start([calculator.py]) --> Q{Kaise chalaya ja raha hai?}
    Q -->|"python calculator.py\n(directly run)"| A["__name__ = '__main__'"]
    Q -->|"import calculator\n(kisi aur file se)"| B["__name__ = 'calculator'"]
    A --> A1["if __name__ == '__main__' block CHALEGA"]
    B --> B1["if __name__ == '__main__' block NAHI CHALEGA"]
```

> **Kyun zaroori hai:** Isse tum ek hi file mein **reusable functions bhi rakh sakte ho aur uska apna test/demo code bhi** — jab file import ho to sirf functions milein, test code na chale.

## 1.6 Python Module ko Dhoondta Kaise Hai — `sys.path`

Jab tum `import calculator` likhte ho, Python ko pata kaise chalta hai ki ye file kahaan hai? Python ek **specific order** mein jagah dhoondta hai, jo `sys.path` mein defined hoti hai:

```python
import sys
print(sys.path)
# ['', '/usr/lib/python3.x', '/usr/lib/python3.x/site-packages', ...]
```

```mermaid
flowchart TD
    Start(["import calculator"]) --> S1["1️⃣ Current directory\n(jahan se script chal rahi hai)"]
    S1 -->|Nahi mila| S2["2️⃣ PYTHONPATH environment variable\n(agar set hai)"]
    S2 -->|Nahi mila| S3["3️⃣ Standard library directories\n(math, os, json, etc.)"]
    S3 -->|Nahi mila| S4["4️⃣ site-packages\n(pip se install kiye hue packages)"]
    S4 -->|Kahin nahi mila| Error["❌ ModuleNotFoundError"]
    S1 -->|Mil gaya| Found["✅ Module load ho gaya"]
    S2 -->|Mil gaya| Found
    S3 -->|Mil gaya| Found
    S4 -->|Mil gaya| Found
```

> Isi wajah se agar tumne apni file ka naam `math.py` rakh diya, to woh built-in `math` module ko **shadow (overshadow)** kar degi, kyunki current directory pehle check hoti hai — aur phir `import math` karne pe tumhari khud ki (adhoori) file load ho jaayegi, asli wala nahi.

## 1.7 Module Caching aur `importlib.reload()`

Python ek smart cheez karta hai: agar ek module **ek baar import ho chuka hai**, to dobara `import` karne pe wo **dobara se poora nahi chalta** — Python `sys.modules` naam ki ek dictionary mein cache rakhta hai.

```python
import calculator   # pehli baar — poori file chalti hai
import calculator   # dusri baar — cache se milta hai, file dobara nahi chalti
```

```mermaid
sequenceDiagram
    participant Code as Tumhara Code
    participant Cache as sys.modules (cache)
    participant File as calculator.py

    Code->>Cache: import calculator (pehli baar)
    Cache->>File: File load karo aur run karo
    File-->>Cache: Module object store ho gaya
    Cache-->>Code: Module return hua

    Code->>Cache: import calculator (dusri baar)
    Cache-->>Code: Seedha cache se return (file dobara NAHI chali)
```

Agar tumne module ki file mein **change** kiya hai aur usko dobara load karwana hai (bina program restart kiye — jaise Jupyter notebooks mein), to:

```python
import importlib
import calculator

importlib.reload(calculator)   # ab file dobara se chalegi, latest changes ke saath
```

## 1.8 `__all__` Variable

Jab koi `from module import *` karta hai, to `__all__` variable control karta hai ki **kya-kya export hoga**.

```python
# calculator.py
__all__ = ["add", "subtract"]   # sirf ye dono hi 'import *' se milenge

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def _internal_helper():   # underscore se start = "private" convention
    pass
```

```python
from calculator import *
add(2, 3)          # ✅ kaam karega
_internal_helper()  # ❌ NameError — __all__ mein nahi tha
```

## 1.9 `__pycache__` aur `.pyc` Files

Jab tum kisi module ko import karte ho, Python usko pehle **compile** karta hai bytecode mein (`.pyc` file), aur usko `__pycache__` folder mein save kar deta hai — taaki **agli baar dobara compile na karna pade** (agar file change nahi hui hai).

```mermaid
flowchart LR
    A["calculator.py\n(source code)"] -->|Pehli baar import| B["Python compile karta hai"]
    B --> C["__pycache__/calculator.cpython-311.pyc\n(bytecode - cached)"]
    D["Agli baar import"] -->|.py file change nahi hui| E["Seedha .pyc se load\n(fast — compile skip)"]
    D -->|.py file change hui| B
```

> Ye purely ek **performance optimization** hai — tumhe manually kuch nahi karna, Python khud handle karta hai. `__pycache__` folder ko `.gitignore` mein daal dena chahiye (Git mein commit nahi karna).

# Part 2: Packages

## 2.1 Package kya hota hai

Jaise multiple functions ko ek module (`.py` file) mein group karte hain, waise hi **multiple modules ko ek folder mein group karna** **package** kehlata hai.

**Real life analogy:** Module ek **kitab** hai. Package ek **almirah** hai jisme related kitabein rakhi hain (jaise sab math ki kitabein ek shelf mein).

```mermaid
flowchart TD
    Package["📦 mypackage/ (Package = folder)"] --> M1["📄 __init__.py"]
    Package --> M2["📄 math_utils.py (Module)"]
    Package --> M3["📄 string_utils.py (Module)"]
    Package --> Sub["📁 helpers/ (Subpackage)"]
    Sub --> S1["📄 __init__.py"]
    Sub --> S2["📄 validators.py"]
```

---

## 2.2 `__init__.py` ka Role

`__init__.py` ek special file hai jo Python ko batati hai ki **"ye folder ek regular folder nahi, ek Python package hai"**. Ye file khaali bhi ho sakti hai, ya usme package-level code bhi ho sakta hai.

```
mypackage/
├── __init__.py
├── math_utils.py
└── string_utils.py
```

Jab bhi tum `import mypackage` karte ho, `__init__.py` ke andar ka code **sabse pehle chalta hai** — ye package ka "welcome/setup" code hota hai.

> **Note:** Python 3.3+ mein `__init__.py` **technically optional** hai (dekho [Section 2.7](#27-namespace-packages-advanced)), par isko rakhna best practice hai kyunki isse package ka structure clear rehta hai aur tum package-level imports control kar sakte ho.

---

## 2.3 Apna Package Banana

Chalo ek real example banate hain:

```
mypackage/
├── __init__.py
├── math_utils.py
└── string_utils.py
```

```python
# math_utils.py
def add(a, b):
    return a + b
```

```python
# string_utils.py
def shout(text):
    return text.upper() + "!"
```

```python
# __init__.py (khaali bhi ho sakti hai)
```

**Ab dusri file se use karna** (assuming `mypackage` folder usi directory mein hai jahan tum script chala rahe ho):

```python
# main.py
from mypackage import math_utils, string_utils

print(math_utils.add(2, 3))          # 5
print(string_utils.shout("hello"))   # HELLO!

# Ya specific function directly import karo
from mypackage.math_utils import add
print(add(10, 20))   # 30
```

## 2.4 Absolute vs Relative Imports

Jab package ke **andar ka ek module doosre module ko import** karna chahta hai, to do tareeke hain:

### Absolute Import (poora path likhna) — recommended

```python
# mypackage/string_utils.py
from mypackage.math_utils import add   # poora path root se
```

### Relative Import (`.` se current package ke relative)

```python
# mypackage/string_utils.py
from .math_utils import add    # '.' matlab "isi package ke andar"
from . import math_utils        # poora module import karna
from ..otherpackage import something   # '..' matlab ek level upar
```

```mermaid
flowchart LR
    subgraph "Absolute Import"
    A1["from mypackage.math_utils import add"] --> A2["Root se poora path\nClear, but lamba"]
    end
    subgraph "Relative Import"
    B1["from .math_utils import add"] --> B2["'.' = current package\nChhota, par sirf package ke andar chalta hai"]
    end
```

> **Rule of thumb:** Package ke andar-andar ke imports ke liye relative imports common hain, par **top-level script mein relative import kabhi nahi chalega** (`ImportError: attempted relative import with no known parent package`) — relative imports sirf tab kaam karte hain jab file **package ke through** import ho rahi ho, directly run nahi ho rahi ho.

## 2.6 `__init__.py` se API Expose Karna

Ek bahut useful pattern — `__init__.py` ka use karke package ka **clean public interface** banana, taaki users ko internal file structure jaanne ki zaroorat na pade.

```python
# mypackage/__init__.py
from .math_utils import add, subtract
from .string_utils import shout

__all__ = ["add", "subtract", "shout"]
```

Ab user directly ye likh sakta hai (bina ye jaane ki `add` kis file mein hai):

```python
from mypackage import add, subtract, shout   # seedha package se, file ka naam pata hone ki zaroorat nahi

print(add(2, 3))
```

Yehi pattern real-world libraries (jaise `numpy`, `pandas`) use karti hain — jab tum `import numpy as np` karte ho, to `np.array()` seedha available hota hai, tumhe pata nahi hota ye internally kis file mein defined hai.

## 2.7 Namespace Packages (Advanced)

Python 3.3 se ek naya concept aaya — **namespace packages**. Ye packages hain jinme `__init__.py` file **hoti hi nahi**, phir bhi Python unko package ki tarah treat kar leta hai.

```
mynamespacepkg/        ← koi __init__.py nahi hai
└── module_a.py
```

**Kab use hota hai:** Jab ek hi logical package ka code **multiple alag-alag locations/directories** mein split ho (jaise bade enterprise projects mein, ya plugin systems mein), tab namespace packages Python ko unhe automatically ek saath jodne dete hain — bina explicit `__init__.py` ke.

> **Beginner ke liye:** Abhi ke liye bas itna samajh lo ki `__init__.py` na ho to bhi Python 3.3+ mein package kaam kar sakta hai, par apne normal projects mein `__init__.py` rakhna hi best practice hai — clarity aur control dono milte hain.

## 2.8 Third-party Packages — `pip` aur PyPI

**PyPI (Python Package Index)** ek **online repository** hai jaha duniya bhar ke developers apne packages publish karte hain (jaise `requests`, `flask`, `numpy`). **`pip`** wo tool hai jisse tum in packages ko apne system pe install karte ho.

```bash
pip install requests          # ek package install karna
pip install requests==2.31.0  # specific version install karna
pip uninstall requests        # uninstall karna
pip list                       # installed packages dekhna
pip show requests              # ek package ki details dekhna
```

**`requirements.txt`** — project ki saari dependencies ek jagah likhna, taaki koi bhi easily setup kar sake:

```
requests==2.31.0
flask==3.0.0
numpy>=1.24.0
```

```bash
pip install -r requirements.txt   # sab ek saath install ho jaate hain
pip freeze > requirements.txt      # current installed packages ki list file mein save karna
```

```mermaid
flowchart LR
    Dev["Developer duniya bhar mein"] -->|Publish karta hai| PyPI[("🌐 PyPI\n(Python Package Index)")]
    PyPI -->|pip install requests| You["Tumhara system"]
    You -->|import requests| Code["Tumhara code"]
```

## 2.9 Virtual Environments

Problem: agar tum multiple projects pe kaam karte ho, aur ek project ko `flask 2.0` chahiye jabki doosre ko `flask 3.0` chahiye — dono ek hi system-wide Python mein install nahi kar sakte, conflict ho jaayega.

**Solution: Virtual Environment (venv)** — har project ke liye ek **isolated, separate Python environment** banana, jisme uski apni packages hoti hain, doosre projects se independent.

```bash
python -m venv myenv          # naya virtual environment banao

# Activate karna:
myenv\Scripts\activate         # Windows
source myenv/bin/activate      # Mac/Linux

pip install flask              # ab ye sirf isi venv mein install hoga

deactivate                     # venv se bahar aana
```

```mermaid
flowchart TD
    System["💻 System-wide Python"] --> P1["Project A venv\nflask==2.0, requests==2.28"]
    System --> P2["Project B venv\nflask==3.0, requests==2.31"]
    System --> P3["Project C venv\ndjango==5.0"]

    style P1 fill:#cce5ff
    style P2 fill:#fff3cd
    style P3 fill:#e6ccff
```

> **Best practice:** Har naye project ke liye naya virtual environment banao. Isse tumhare projects ek doosre ke dependencies ko kabhi disturb nahi karenge, aur jab tum project kisi aur ko share karoge, `requirements.txt` se woh apna khud ka clean environment bana sakta hai.

## 2.11 Common Pitfalls

### a) Circular Imports

Jab do modules **ek doosre ko import** karne ki koshish karte hain:

```python
# a.py
import b
def func_a():
    b.func_b()
```

```python
# b.py
import a    # ❌ circular! a already b ko import kar raha hai
def func_b():
    a.func_a()
```

**Fix:** Import ko function ke andar move karo (lazy import), ya code ko restructure karke shared logic ko third module mein nikaal lo.

### b) File ka naam built-in module se clash karna

```python
# ❌ Apni file ka naam "random.py" rakh dena
# Ab jab tum 'import random' karoge (built-in ke liye), Python tumhari file utha lega!
```

**Fix:** Kabhi bhi apni files ka naam built-in modules jaisa mat rakho (`math.py`, `random.py`, `json.py`, etc.)

### c) `__init__.py` bhoolna (purane Python mein zaroori)

Agar package import nahi ho raha aur error aa raha hai, check karo ki `__init__.py` file exist karti hai ya nahi.

```mermaid
flowchart TD
    A["ModuleNotFoundError\nya ImportError aaya?"] --> B{Kya check karein?}
    B --> C["1. File ka naam kisi built-in\nmodule se clash to nahi kar raha?"]
    B --> D["2. __init__.py file hai package mein?"]
    B --> E["3. Current directory se import\ncorrect path pe hai?"]
    B --> F["4. Circular import to nahi ho raha?"]
```
