# Comprehensive Guide to Python Functions: Beginner to Advanced

This document outlines everything you need to know about Python functions, progressing from fundamental concepts to highly advanced, production-level paradigms.

---

## 1. Beginner Level

### 1.1 What is a Function?
A function is a reusable block of organized code used to perform a single, related action. Functions provide better modularity for your application and a high degree of code reusability.

### 1.2 Basic Syntax
In Python, functions are defined using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`.

```python
def greet():
    """This is a simple docstring explaining the function."""
    print("Hello, Welcome to Python Functions!")

# Calling the function
greet()
```

### 1.3 Parameters and Arguments
* **Parameters**: Variables listed inside the function definition parentheses.
* **Arguments**: The actual values sent to the function when it is called.

```python
def greet_user(username):  # username is a parameter
    print(f"Hello, {username}!")

greet_user("Alice")       # "Alice" is an argument
```

### 1.4 Return Statement
The `return` statement exits a function and optionally passes back an expression or value to the caller. If no return statement is specified, the function returns `None`.

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 7)
print(result)  # Output: 12
```

---

## 2. Intermediate Level

### 2.1 Argument Types
Python supports flexible ways to pass arguments to functions.

#### Default Arguments
You can assign default values to parameters. If no argument is passed during the function call, the default value is used.

```python
def display_info(name, role="Developer"):
    print(f"Name: {name}, Role: {role}")

display_info("Bob")                 # Uses default role
display_info("Charlie", "Manager")  # Overrides default role
```

#### Keyword Arguments
Arguments can be explicitly passed using the parameter name, making the order irrelevant.

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type} named {pet_name}.")

describe_pet(pet_name="Whiskers", animal_type="cat")
```

#### Arbitrary Arguments (`*args` and `**kwargs`)
* `*args`: Receives a tuple of positional arguments when the total count is unknown.
* `**kwargs`: Receives a dictionary of keyword arguments when the total count is unknown.

```python
def total_sum(*args):
    return sum(args)

print(total_sum(1, 2, 3, 4, 5))  # Output: 15

def print_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_profile(name="John", age=30, city="New York")
```

### 2.2 Docstrings and Annotations
Type hinting helps linting tools and developers understand what data types a function expects.

```python
def calculate_area(radius: float) -> float:
    """
    Calculates the area of a circle.
    
    Parameters:
    radius (float): The radius of the circle.
    
    Returns:
    float: The calculated area.
    """
    import math
    return math.pi * (radius ** 2)
```

### 2.3 Scope and Lifetime of Variables
* **Local Scope**: Variables declared inside a function are accessible only within that function.
* **Global Scope**: Variables declared outside functions are globally accessible. Use the `global` keyword to modify them inside a function.

---

## 3. Advanced Level

### 3.1 First-Class Citizens & Higher-Order Functions
In Python, functions are first-class objects. This means they can be passed as arguments, returned from other functions, and assigned to variables.

```python
def square(x):
    return x * x

def apply_function(func, value):
    return func(value)

print(apply_function(square, 5))  # Output: 25
```

### 3.2 Closures
A closure is a function object that remembers values in enclosing scopes even if they are not present in memory.

```python
def outer_multiplier(factor):
    def inner_multiply(number):
        return number * factor
    return inner_multiply

double = outer_multiplier(2)
print(double(10))  # Output: 20
```

### 3.3 Decorators
Decorators wrap another function to extend its behavior without permanently modifying it.

```python
def log_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Executing: {func.__name__}")
        result = func(*args, **kwargs)
        print("Execution completed.")
        return result
    return wrapper

@log_decorator
def complex_task():
    print("Performing task...")

complex_task()
```

### 3.4 Lambda (Anonymous) Functions
Small anonymous functions can be created using the `lambda` keyword. They are restricted to a single expression.

```python
multiply = lambda x, y: x * y
print(multiply(4, 5))  # Output: 20
```

---

## 4. Master / Expert Level

### 4.1 Generators and the `yield` Keyword
Generators permit you to declare a function that behaves like an iterator, producing a sequence of values lazily over time using `yield`.

```python
def infinite_sequence():
    num = 0
    while True:
        yield num
        num += 1

gen = infinite_sequence()
print(next(gen))  # 0
print(next(gen))  # 1
```

### 4.2 Positional-Only and Keyword-Only Arguments
You can strictly enforce how arguments are passed using `/` and `*`.
* Arguments before `/` must be **positional-only**.
* Arguments after `*` must be **keyword-only**.

```python
def strict_function(pos_only, /, mixed, *, kw_only):
    print(pos_only, mixed, kw_only)

strict_function(1, 2, kw_only=3)          # Valid
strict_function(1, mixed=2, kw_only=3)    # Valid
# strict_function(pos_only=1, mixed=2, kw_only=3) # TypeError
```

### 4.3 Recursive Functions
Functions can call themselves. Ensure a proper base case exists to avoid `RecursionError`.

```python
def factorial(n: int) -> int:
    if n <= 1:
        return 1
    return n * factorial(n - 1)
```

### 4.4 Asynchronous Functions (`async` / `await`)
Asynchronous functions run concurrently via event loops, widely applied in modern web-framework pipelines and I/O heavy processes.

```python
import asyncio

async def fetch_data():
    print("Fetching data...")
    await asyncio.sleep(2)
    print("Data retrieved!")

# To run: asyncio.run(fetch_data())
```

# Memory Mein Function Kaise Execute Hota Hai? (Deep Dive)

Jab bhi tumhara Python program run hota hai, toh OS (Operating System) tumhare program ko RAM mein ek jagah deta hai jise **Process Memory** kehte hain. Is memory ke do sabse important hisse hote hain:
1. **Stack Memory (Fast & Structured)**
2. **Heap Memory (Flexible & Dynamic)**

---

## 1. Stack aur Heap ka Asli Matlab

| Memory Area | Real-life Analogy | Python mein iska kaam | Speed |
| :--- | :--- | :--- | :--- |
| **Stack** | Plates ki dheri (LIFO) | Track rakhta hai ki abhi kaunsa function chal raha hai aur variables ke names/references kya hain. | **Super Fast** ⚡ |
| **Heap** | Ek bada khula maidan | Jitne bhi actual objects, lists, tuples, ya integers bante hain, woh yahan store hote hain. | Thoda Slow |

---

## 2. Execution Context aur Stack Frame 🖼️

Jab bhi tum kisi function ko **call** karte ho, toh Stack Memory ke andar us function ke naam ka ek **"Stack Frame"** ban jata hai. Isme 3 main cheezein hoti hain:
* **Local Variables & Parameters**
* **Return Address** (Khatam hone ke baad code ko wapas kis line par jaana hai)

> 💡 **Golden Rule:** Jaise hi function apna kaam khatam karta hai (`return` hota hai), uska Stack Frame stack se **pop (delete)** ho jata hai.

---

## 3. Step-by-Step Code Execution (Dry Run)

```python
def square(x):
    result = x * x
    return result

def main():
    num = 5
    ans = square(num)
    print(ans)

main()
```

1. **Step 1:** `main()` call hua -> Stack ke upar `main()` ka frame bana. Usme `num = 5` save hua.
2. **Step 2:** `main()` ke andar se `square(num)` call hua -> `main()` ke upar `square()` ka naya frame bana. Parameter `x = 5` aur local variable `result = 25` store hua.
3. **Step 3:** `square()` khatam hua -> `square()` ka frame Stack se poori tarah **delete** ho gaya. Control wapas `main()` ke paas aaya aur `ans = 25` save hua.
4. **Step 4:** `main()` khatam hua -> `main()` ka frame bhi Stack se saaf! Memory ekdum khaali.

---

## 4. Advance Concept: Recursion aur Stack Overflow 🚨

Jab ek function apne aap ko hi baar-baar call karta hai bina ruke, toh stack frames ek ke upar ek bante chale jaate hain. Ek waqt par RAM ki Stack Memory bhar jaati hai, jise **Stack Overflow** kehte hain. Python isse bachne ke liye `RecursionError` de deta hai.

## 3. Advanced Level (Pro)

### Lambda Functions (Anonymous Functions)
Yeh ek line ke chote functions hote hain jinka koi naam nahi hota. Ismein `def` ki jagah `lambda` keyword ka use hota hai.

```python
# Syntax: lambda parameters : expression
square = lambda x: x * x
print(square(4)) # Output: 16

# Do numbers ko add karna
add = lambda a, b: a + b
print(add(3, 7)) # Output: 10
```
* **Anonymous (Nameless):** They do not use the standard `def` keyword and are not bound to a specific identifier name upon creation unless explicitly assigned to a variable.
* **Single Expression Only:** The body can only contain a single expression. It cannot contain multi-line logic, loops (`for`/`while`), or complex control structures.
* **Implicit Return:** You do not use the `return` keyword. The single expression inside the body is automatically evaluated and returned.
* **Inline Syntax:** They are written entirely on a single line of code, making them ideal for quick, throwaway operations.
* **Accepts Multiple Arguments:** Just like normal functions, a lambda can take zero or more arguments, but it executes only one expression using those arguments.
* **No Docstrings or Annotations:** Because of their strict single-line syntax, they cannot include documentation strings (docstrings) or type hints in their definition.

### Basic Syntax
```python
lambda argument1, argument2: expression
```

### Basic Example
```python
# A lambda function that adds two numbers
add = lambda x, y: x + y
print(add(5, 3))  # Output: 8
```

## Part 2: Higher-Order Functions (HOFs)

A **Higher-Order Function** is a function that does at least one of the following:
1. Takes one or more functions as **arguments**.
2. Returns a function as its **result**.

### Characteristics of HOFs
* **First-Class Citizens:** They treat functions like any other object (like integers or strings) that can be passed around.
* **Code Reusability:** They separate the "looping/traversal" logic from the "action/transformation" logic.

---
```python
def square(x):
    return x**2

def transform(f,L):
    output=[]
    for i in L:
        output.append(f(i))
    print(output)

L=[1,2,3,4,5]

transform(square,L)
```
## Part 3: Usability of Lambda Functions with Higher-Order Functions

Lambda functions are most powerful when used **inside Higher-Order Functions**. Because HOFs often require a quick, one-time operation, writing a full `def` function is unnecessary. Passing a lambda keeps the code clean and contained.

Here are the most common built-in Higher-Order Functions in Python used with lambdas:

### 1. With `map()`
The `map()` function applies a transformation function to every item in an iterable.

```python
numbers = [1, 2, 3, 4, 5]

# Using lambda to square each number
squared = list(map(lambda x: x**2, numbers))
print(squared)  # Output: [1, 4, 9, 16, 25]
```

### 2. With `filter()`
The `filter()` function extracts elements from an iterable for which a function returns `True`.

```python
numbers = [1, 2, 3, 4, 5, 6]

# Using lambda to filter out only even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)  # Output: [2, 4, 6]
```

### 3. With `sorted()` or `.sort()`
The `sorted()` function can take a `key` argument (which expects a function) to determine custom sorting logic.

```python
# A list of tuples containing (Name, Age)
students = [("Alice", 25), ("Bob", 20), ("Charlie", 23)]

# Sort the students by their Age (index 1 of the tuple) using lambda
sorted_students = sorted(students, key=lambda student: student[1])
print(sorted_students)  # Output: [('Bob', 20), ('Charlie', 23), ('Alice', 25)]
```
### 4. With `reduce()`
The `reduce()` function applies a rolling computation to sequential pairs of values in an iterable. It processes the first two elements, takes that result, combines it with the third element, and repeats this until only a single value remains.

*Note: You must import `reduce` from the `functools` module.*

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Using lambda to multiply all numbers together (1 * 2 * 3 * 4 * 5)
product = reduce(lambda x, y: x * y, numbers)
print(product)  # Output: 120

functools.reduce(lambda x,y:x if x<y else y,[23,11,45,10,1])
#finds min from the list
```

#### How it works step-by-step:
1. `x` takes `1`, `y` takes `2` $\rightarrow$ returns `1 * 2 = 2`
2. `x` takes the previous result `2`, `y` takes `3` $\rightarrow$ returns `2 * 3 = 6`
3. `x` takes `6`, `y` takes `4` $\rightarrow$ returns `6 * 4 = 24`
4. `x` takes `24`, `y` takes `5` $\rightarrow$ returns `24 * 5 = 120`


### 4. Custom Higher-Order Function Example
You can also build your own HOF that accepts a lambda function.

```python
# Custom HOF that applies any math operation twice
def apply_twice(func, value):
    return func(func(value))

# Passing a lambda that adds 5
result = apply_twice(lambda x: x + 5, 10)
print(result)  # Output: 20 (10 + 5 + 5)

## 4. Master Level (Expert)

### Generators (`yield` keyword)
Normal function ek hi baar mein poori value `return` karke khatam ho jata hai. Lekin Generator ek-ek karke value bhejta hai (`yield` use karke) aur apni state yaad rakhta hai. Isse memory bohot bachti hai.

```python
def count_up_to(max_num):
    count = 1
    while count <= max_num:
        yield count # Value pause karke bahar bhejta hai
        count += 1

counter = count_up_to(3)
print(next(counter)) # Output: 1
print(next(counter)) # Output: 2
print(next(counter)) # Output: 3
```

### Positional-Only (`/`) aur Keyword-Only (`*`) Arguments
* `/` se pehle wale arguments sirf position se pass ho sakte hain (naam nahi likh sakte).
* `*` ke baad wale arguments sirf naam (keyword) se hi pass ho sakte hain.

```python
def strict_func(a, /, b, *, c):
    print(a, b, c)

# strict_func(a=1, b=2, c=3) # Error! 'a' positional-only hai.
# strict_func(1, 2, 3)       # Error! 'c' keyword-only hai.

strict_func(1, 2, c=3) # Ekdum Sahi!
```
